**Preprocesamiento**

```python
import re

def basic_clean(text: str) -> str:
    if pd.isna(text):
        return text

    text = str(text).strip()

    # Normalizar espacios múltiples
    text = re.sub(r'\s+', ' ', text)

    # Quitar espacio antes de signos de puntuación
    text = re.sub(r'\s+([,.;:?!])', r'\1', text)

    # Quitar espacio después de signos de apertura
    text = re.sub(r'([¿¡(])\s+', r'\1', text)

    # Quitar espacio antes de cerrar paréntesis
    text = re.sub(r'\s+\)', ')', text)

    # Asegurar espacio después de coma, punto y coma, etc. si va seguida de letra o apertura
    text = re.sub(r'([,.;])([A-Za-zÁÉÍÓÚáéíóúÑñ¿¡«(])', r'\1 \2', text)
    text = re.sub(r'([?!])([A-Za-zÁÉÍÓÚáéíóúÑñ¿¡«(])', r'\1 \2', text)

    # Quitar espacio antes de :
    text = re.sub(r'\s+:', ':', text)

    # Asegurar un espacio después de :
    text = re.sub(r':(?=\S)', ': ', text)

    # Reducir posibles dobles espacios creados por la regla anterior
    text = re.sub(r':\s+', ': ', text)

    # Quitar espacio incorrecto antes de comillas de cierre
    text = re.sub(r'([?!.,;])\s+(")', r'\1\2', text)

    # Corregir erratas obvias y sistemáticas
    text = re.sub(r'\bSe prohibe\b', 'Se prohíbe', text)
    text = re.sub(r'\bse prohibe\b', 'se prohíbe', text)

    # Eliminar puntos finales repetidos
    text = re.sub(r'\.+$', '.', text)

    # Mayúscula inicial básica
    if text:
        text = text[0].upper() + text[1:]

    # Añadir punto final si no termina en cierre válido
    if text and text[-1] not in '.?!»"':
        text += '.'

    return text

def normalizar_dataframe(df: pd.DataFrame, col_input: str, col_target: str) -> pd.DataFrame:
    df = df[[col_input, col_target]].copy()

    # Eliminar NaN
    df = df.dropna(subset=[col_input, col_target])

    # Limpiar ambas columnas
    df[col_input] = df[col_input].apply(basic_clean)
    df[col_target] = df[col_target].apply(basic_clean)

    # Eliminar filas vacías tras la limpieza
    df = df[
        (df[col_input].str.strip() != '') &
        (df[col_target].str.strip() != '')
    ]

    # Eliminar duplicados exactos
    df = df.drop_duplicates()

    # Resetear índice
    df = df.reset_index(drop=True)

    return df

def normalizar_dataframe_grande(df: pd.DataFrame, col_input: str, col_target: str, col_changed: str) -> pd.DataFrame:
    df = df[[col_input, col_target, col_changed]].copy()

    # Eliminar NaN
    df = df.dropna(subset=[col_input, col_target, col_changed])

    # Limpiar ambas columnas
    df[col_input] = df[col_input].apply(basic_clean)
    df[col_target] = df[col_target].apply(basic_clean)
    df[col_changed] = df[col_changed].apply(basic_clean)

    # Eliminar filas vacías tras la limpieza
    df = df[
        (df[col_input].str.strip() != '') &
        (df[col_target].str.strip() != '') &
        (df[col_changed].str.strip() != '')
    ]

    # Eliminar duplicados exactos
    df = df.drop_duplicates()

    # Resetear índice
    df = df.reset_index(drop=True)

    return df

def normalizar_columna_simple(df: pd.DataFrame, col: str) -> pd.DataFrame:
    df = df[[col]].copy()

    df = df.dropna(subset=[col])

    df[col] = df[col].apply(basic_clean)

    df = df[df[col].str.strip() != '']

    df = df.drop_duplicates().reset_index(drop=True)

    return df
```

```python
df_sintetico_final = normalizar_dataframe_grande(
    df_sintetico_final,
    'input',
    'target',
    'changed',
)
```

```python
df_noticias = normalizar_dataframe(
    df_noticias,
    'input',
    'target'
)

df_noticias2 = normalizar_dataframe(
    df_noticias2,
    'input',
    'target'
)

df_noerror = normalizar_dataframe(
    df_noerror,
    'input',
    'target'
)
```

```python
df_noticias_extendido = pd.concat(
    [
        df_noerror,
        df_noticias,
        df_noticias2
    ],
    ignore_index=True
).drop_duplicates().reset_index(drop=True)
```

```python
df_sintetico_final = df_sintetico_final.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

df_noticias_final = df_noticias_extendido.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

**Evaluación de modelos**

# **BiLSTM-LSTM**

```python
# ============================================================
# PIPELINE GEC — BiLSTM-LSTM
# ============================================================

import json, time, gc, random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence
from transformers import AutoTokenizer
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
# ============================================================

SEED            = 42
TOKENIZER_NAME  = "bert-base-multilingual-cased"
MAX_LEN         = 128
BATCH_SIZE      = 16
EMB_SIZE        = 256
HIDDEN_SIZE     = 256
N_LAYERS        = 2
DROPOUT         = 0.3
LR              = 1e-3
WEIGHT_DECAY    = 1e-5
N_EPOCHS        = 256
PATIENCE        = 30
MAX_GRAD_NORM   = 1.0
BEST_MODEL_PATH = "best_model_bilstm.pt"

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. TOKENIZADOR
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
PAD_ID = tokenizer.pad_token_id
CLS_ID = tokenizer.cls_token_id
SEP_ID = tokenizer.sep_token_id

print(f"Vocab size: {tokenizer.vocab_size}")

# ============================================================
# 3. DATASET Y DATALOADERS
# ============================================================

class GECDataset(Dataset):
    def __init__(self, df, tokenizer, max_len=128):
        self.src_texts = df["input"].tolist()
        self.tgt_texts = df["target"].tolist()
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.src_texts)

    def __getitem__(self, idx):
        src_enc = self.tokenizer.encode(
            self.src_texts[idx], max_length=self.max_len,
            truncation=True, add_special_tokens=True,
        )
        tgt_enc = self.tokenizer.encode(
            self.tgt_texts[idx], max_length=self.max_len,
            truncation=True, add_special_tokens=True,
        )
        return (
            torch.tensor(src_enc, dtype=torch.long),
            torch.tensor(tgt_enc, dtype=torch.long),
        )

def collate_fn(batch):
    srcs, tgts = zip(*batch)
    return (
        pad_sequence(srcs, batch_first=True, padding_value=PAD_ID),
        pad_sequence(tgts, batch_first=True, padding_value=PAD_ID),
    )

train_loader = DataLoader(
    GECDataset(df_train, tokenizer, MAX_LEN),
    batch_size=BATCH_SIZE, shuffle=True,
    collate_fn=collate_fn, pin_memory=True,
)
val_loader = DataLoader(
    GECDataset(df_val, tokenizer, MAX_LEN),
    batch_size=BATCH_SIZE, shuffle=False,
    collate_fn=collate_fn, pin_memory=True,
)
test_loader = DataLoader(
    GECDataset(df_test, tokenizer, MAX_LEN),
    batch_size=1, shuffle=False, collate_fn=collate_fn,
)
noticias_loader = DataLoader(
    GECDataset(df_noticias_final, tokenizer, MAX_LEN),
    batch_size=1, shuffle=False, collate_fn=collate_fn,
)

print(f"Train: {len(train_loader)} batches | Val: {len(val_loader)} batches")
print(f"Test: {len(test_loader)} | Noticias: {len(noticias_loader)}")

# ============================================================
# 4. MODELO
# ============================================================

class Encoder(nn.Module):
    def __init__(self, vocab_size, emb_size, hidden_size, n_layers, dropout, pad_id):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_size, padding_idx=pad_id)
        self.rnn = nn.LSTM(
            emb_size, hidden_size, num_layers=n_layers,
            batch_first=True, bidirectional=True,
            dropout=dropout if n_layers > 1 else 0.0,
        )
        self.dropout = nn.Dropout(dropout)
        self.fc_h = nn.Linear(hidden_size * 2, hidden_size)
        self.fc_c = nn.Linear(hidden_size * 2, hidden_size)

    def forward(self, src):
        embedded = self.dropout(self.embedding(src))
        _, (hidden, cell) = self.rnn(embedded)
        h = torch.tanh(self.fc_h(torch.cat((hidden[-2], hidden[-1]), dim=1)))
        c = torch.tanh(self.fc_c(torch.cat((cell[-2],   cell[-1]),   dim=1)))
        return h.unsqueeze(0), c.unsqueeze(0)

class Decoder(nn.Module):
    def __init__(self, vocab_size, emb_size, hidden_size, dropout, pad_id):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_size, padding_idx=pad_id)
        self.rnn       = nn.LSTM(emb_size, hidden_size, batch_first=True)
        self.fc_out    = nn.Linear(hidden_size, vocab_size)
        self.dropout   = nn.Dropout(dropout)

    def forward(self, tgt, hidden, cell):
        emb = self.dropout(self.embedding(tgt))
        out, _ = self.rnn(emb, (hidden, cell))
        return self.fc_out(out)

class Seq2SeqGEC(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, tgt):
        hidden, cell = self.encoder(src)
        return self.decoder(tgt, hidden, cell)

encoder = Encoder(tokenizer.vocab_size, EMB_SIZE, HIDDEN_SIZE, N_LAYERS, DROPOUT, PAD_ID)
decoder = Decoder(tokenizer.vocab_size, EMB_SIZE, HIDDEN_SIZE, DROPOUT, PAD_ID)
model   = Seq2SeqGEC(encoder, decoder).to(device)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parámetros entrenables: {n_params:,}")

# ============================================================
# 5. OPTIMIZADOR Y CRITERIO
# ============================================================

optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=15
)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_ID)

# ============================================================
# 6. MÉTRICAS GEC
# ============================================================

def decode_clean(token_ids):
    if isinstance(token_ids, torch.Tensor):
        token_ids = token_ids.cpu().tolist()
    filtered = [
        t for t in token_ids
        if t not in (PAD_ID, CLS_ID, SEP_ID, tokenizer.unk_token_id)
        and t is not None
    ]
    return tokenizer.decode(
        filtered, skip_special_tokens=True,
        clean_up_tokenization_spaces=True
    ).strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    """
    Métricas a dos niveles:

    1. Clasificación binaria (¿necesita corrección?):
       - Positivo  = src != tgt  → la frase tiene errores y necesita corrección
       - Negativo  = src == tgt  → la frase es correcta, no hay que tocarla
       - Predicción positiva = el modelo modifica la frase (pred != src)
       - Predicción negativa = el modelo NO modifica la frase (pred == src)

       TP: frase con errores   → modelo la modifica   ✓ correcto
       TN: frase sin errores   → modelo no la toca    ✓ correcto
       FP: frase sin errores   → modelo la modifica   ✗ falsa corrección
       FN: frase con errores   → modelo no la toca    ✗ no corrige

    2. Calidad de la corrección (token-level, solo cuando pred != src):
       GLEU, F0.5, precision, recall, edit distance.
       Se calcula siempre sobre todas las muestras para poder agregar.
    """
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # ── Nivel 1: clasificación binaria ────────────────────
    # "needs_correction" es la etiqueta real (ground truth)
    needs_correction = src_text.strip() != tgt_text.strip()   # True → positivo
    pred_modified    = pred_text.strip() != src_text.strip()  # True → el modelo predice positivo

    # ── Nivel 2: calidad token-level ──────────────────────
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        # clasificación binaria
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        # calidad token-level
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    """
    Agrega métricas de clasificación binaria y token-level.

    Clasificación binaria
    ─────────────────────
    Positivo  = frase CON errores  (needs_correction = True)
    Negativo  = frase SIN errores  (needs_correction = False)
    Pred pos  = modelo modifica    (pred_modified = True)
    Pred neg  = modelo no modifica (pred_modified = False)

      TP = needs_correction=True  & pred_modified=True
      TN = needs_correction=False & pred_modified=False
      FP = needs_correction=False & pred_modified=True   (falsa corrección)
      FN = needs_correction=True  & pred_modified=False  (no corrige cuando debería)

    Accuracy  = (TP + TN) / N
    Precision = TP / (TP + FP)
    Recall    = TP / (TP + FN)
    F0.5      = (1 + 0.5²) · P · R / (0.5² · P + R)   [penaliza más FP que FN]
    """
    if not all_metrics:
        return {}

    n = len(all_metrics)

    # ── Clasificación binaria ──────────────────────────────
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    n_positive = tp + fn   # frases que realmente necesitaban corrección
    n_negative = tn + fp   # frases que eran correctas

    # ── Token-level (calidad de corrección) ───────────────
    gleu_vals      = [m["gleu"]          for m in all_metrics]
    f05_tok_vals   = [m["f05_tok"]       for m in all_metrics]
    prec_tok_vals  = [m["precision_tok"] for m in all_metrics]
    rec_tok_vals   = [m["recall_tok"]    for m in all_metrics]
    ed_norm_vals   = [m["ed_norm"]       for m in all_metrics]

    return {
        # conteos
        "n_total":          n,
        "n_positive":       n_positive,   # con errores (ground truth)
        "n_negative":       n_negative,   # sin errores (ground truth)
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        # clasificación binaria
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        # token-level
        "gleu_mean":        float(np.mean(gleu_vals)),
        "gleu_std":         float(np.std(gleu_vals)),
        "f05_tok_mean":     float(np.mean(f05_tok_vals)),
        "f05_tok_std":      float(np.std(f05_tok_vals)),
        "precision_tok":    float(np.mean(prec_tok_vals)),
        "recall_tok":       float(np.mean(rec_tok_vals)),
        "ed_norm_mean":     float(np.mean(ed_norm_vals)),
        # auxiliar
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }

# ============================================================
# 7. FUNCIÓN DE EVALUACIÓN
# ============================================================

def evaluate_loader(model, loader, criterion, src_texts_list=None,
                    compute_metrics=False):
    model.eval()
    total_loss  = 0.0
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (src, tgt) in enumerate(loader):
            src, tgt = src.to(device), tgt.to(device)

            output = model(src, tgt)
            loss = criterion(
                output[:, 1:].reshape(-1, tokenizer.vocab_size),
                tgt[:, 1:].reshape(-1),
            )
            total_loss += loss.item()

            if compute_metrics:
                pred_ids = output.argmax(dim=-1)

                for b in range(src.size(0)):
                    pred_text = decode_clean(pred_ids[b].cpu().tolist())
                    tgt_text  = decode_clean(tgt[b].cpu().tolist())
                    src_text  = decode_clean(src[b].cpu().tolist())

                    if src_texts_list is not None:
                        idx = i * loader.batch_size + b
                        if idx < len(src_texts_list):
                            src_text = src_texts_list[idx]

                    metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
                    all_metrics.append(metrics)
                    predictions.append({
                        "src": src_text, "tgt": tgt_text, "pred": pred_text,
                        **metrics,
                    })

    avg_loss = total_loss / len(loader)
    agg      = aggregate_metrics(all_metrics) if all_metrics else {}
    return avg_loss, agg, predictions


def print_all_samples(preds, dataset_name):
    """Imprime todas las muestras con su etiqueta TP/TN/FP/FN."""
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if nc and pm:
            label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm:
            label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and pm:
            label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:
            label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, loss, inference_time, n_samples, dataset_name):
    """Imprime todas las métricas agregadas de forma clara."""
    tps = agg.get('throughput_sps', n_samples / inference_time)
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Loss:              {loss:.4f}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Clasificación binaria (¿necesita corrección?) ──")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}")
    print(f"     Precision:  {agg.get('precision', 0):.4f}")
    print(f"     Recall:     {agg.get('recall',    0):.4f}")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}")
    print()
    print(f"  ── Calidad de corrección (token-level) ──")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ──")
    print(f"     Correction rate:   {agg.get('correction_rate',  0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections', 0)}")
    print()
    print(f"  ── Tiempos ──")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 8. ENTRENAMIENTO
# ============================================================

print("\n" + "="*60)
print("ENTRENAMIENTO BiLSTM-LSTM")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

set_seed(SEED)

print("Warmup...")
model.train()
_wb = 0
for src_w, tgt_w in train_loader:
    src_w, tgt_w = src_w.to(device), tgt_w.to(device)
    with torch.no_grad():
        _ = model(src_w, tgt_w)
    _wb += 1
    if _wb >= 3:
        break

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

print("Warmup completado.")

if torch.cuda.is_available():
    torch.cuda.synchronize()
TRAIN_START = time.perf_counter()

best_val_loss     = float("inf")
epochs_no_improve = 0
epoch_log         = []

for epoch in range(1, N_EPOCHS + 1):
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_start = time.perf_counter()

    model.train()
    train_loss = 0.0
    for src, tgt in train_loader:
        src, tgt = src.to(device), tgt.to(device)
        optimizer.zero_grad()
        output = model(src, tgt)
        loss = criterion(
            output[:, 1:].reshape(-1, tokenizer.vocab_size),
            tgt[:, 1:].reshape(-1),
        )
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
        optimizer.step()
        train_loss += loss.item()
    train_loss /= len(train_loader)

    val_loss, _, _ = evaluate_loader(model, val_loader, criterion)
    scheduler.step(val_loss)
    current_lr = optimizer.param_groups[0]['lr']

    if val_loss < best_val_loss:
        best_val_loss     = val_loss
        epochs_no_improve = 0
        torch.save({
            "epoch":            epoch,
            "model_state_dict": model.state_dict(),
            "val_loss":         val_loss,
            "seed":             SEED,
        }, BEST_MODEL_PATH)
    else:
        epochs_no_improve += 1

    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_time = time.perf_counter() - epoch_start

    epoch_log.append({
        "epoch": epoch, "train_loss": train_loss,
        "val_loss": val_loss, "lr": current_lr,
        "epoch_time_s": epoch_time,
    })

    if epoch % 10 == 0 or epoch == 1:
        print(
            f"Epoch {epoch:3d}/{N_EPOCHS} | "
            f"Train: {train_loss:.4f} | Val: {val_loss:.4f} | "
            f"LR: {current_lr:.2e} | Time: {epoch_time:.2f}s"
        )

    if epochs_no_improve >= PATIENCE:
        print(f"\nEarly stopping en epoch {epoch}.")
        break

if torch.cuda.is_available():
    torch.cuda.synchronize()
TRAIN_END        = time.perf_counter()
total_train_time = TRAIN_END - TRAIN_START

print(f"\nMejor val_loss: {best_val_loss:.4f}")
print(f"Tiempo total: {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
print(f"Tiempo promedio por época: {total_train_time/len(epoch_log):.2f}s")

# ============================================================
# 9. CARGA DEL MEJOR MODELO
# ============================================================

checkpoint = torch.load(BEST_MODEL_PATH, map_location=device)
model.load_state_dict(checkpoint["model_state_dict"])
print(f"\nMejor modelo: epoch {checkpoint['epoch']} (val_loss={checkpoint['val_loss']:.4f})")

# ============================================================
# 10. EVALUACIÓN TEST
# ============================================================

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_loss, test_agg, test_preds = evaluate_loader(
    model, test_loader, criterion,
    src_texts_list=df_test["input"].tolist(),
    compute_metrics=True,
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_loss, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 11. EVALUACIÓN NOTICIAS
# ============================================================

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_loss, noticias_agg, noticias_preds = evaluate_loader(
    model, noticias_loader, criterion,
    src_texts_list=df_noticias_final["input"].tolist(),
    compute_metrics=True,
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_loss, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 12. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":      "BiLSTM-LSTM",
    "seed":            SEED,
    "device":          device,
    "n_params":        n_params,
    "batch_size":      BATCH_SIZE,
    "max_len":         MAX_LEN,
    "emb_size":        EMB_SIZE,
    "hidden_size":     HIDDEN_SIZE,
    "n_layers":        N_LAYERS,
    "dropout":         DROPOUT,
    "decode_strategy": "argmax_tf1",

    "epochs_run":         len(epoch_log),
    "best_epoch":         checkpoint["epoch"],
    "best_val_loss":      float(best_val_loss),
    "total_train_time_s": total_train_time,
    "avg_epoch_time_s":   total_train_time / len(epoch_log),

    # ── TEST: clasificación binaria ──
    "test_loss":              test_loss,
    "test_n_total":           test_agg.get("n_total", 0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp", 0),
    "test_tn":                test_agg.get("tn", 0),
    "test_fp":                test_agg.get("fp", 0),
    "test_fn":                test_agg.get("fn", 0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # ── TEST: token-level ──
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # ── NOTICIAS: clasificación binaria ──
    "noticias_loss":           noticias_loss,
    "noticias_n_total":        noticias_agg.get("n_total", 0),
    "noticias_n_positive":     noticias_agg.get("n_positive", 0),
    "noticias_n_negative":     noticias_agg.get("n_negative", 0),
    "noticias_tp":             noticias_agg.get("tp", 0),
    "noticias_tn":             noticias_agg.get("tn", 0),
    "noticias_fp":             noticias_agg.get("fp", 0),
    "noticias_fn":             noticias_agg.get("fn", 0),
    "noticias_accuracy":       noticias_agg.get("accuracy",  0),
    "noticias_precision":      noticias_agg.get("precision", 0),
    "noticias_recall":         noticias_agg.get("recall",    0),
    "noticias_f05":            noticias_agg.get("f05",       0),
    # ── NOTICIAS: token-level ──
    "noticias_gleu_mean":      noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":       noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":   noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":    noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":  noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":     noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":   noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_bilstm_ger.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
        "epoch_log":            epoch_log,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_bilstm_ger.json'")
```

**Metodología:**

Se implementa una arquitectura seq2seq (sequence-to-sequence, donde el modelo transforma una secuencia de entrada en otra de salida) compuesta por un **encoder BiLSTM bidireccional de 2 capas** (red recurrente que lee la frase en ambas direcciones para capturar contexto izquierdo y derecho) y un **decoder LSTM unidireccional** (red recurrente que genera la frase corregida token a token), con embeddings y estado oculto de 256 dimensiones y 95.3M parámetros entrenables. El modelo se entrena desde cero (from scratch, sin preentrenamiento) con el optimizador Adam (lr=1e-3, weight decay=1e-5), gradient clipping (técnica que evita explosión de gradientes, norma máxima 1.0) y early stopping (detención anticipada cuando la pérdida de validación no mejora tras 30 épocas consecutivas) sobre un máximo de 256 épocas. La tokenización subword se realiza con bert-base-multilingual-cased. El entrenamiento utiliza **teacher forcing completo** (TF=1.0), estrategia en la que el decoder recibe en cada paso el token correcto como entrada en lugar de su propia predicción anterior, lo que estabiliza el entrenamiento con datos escasos (~336 muestras). Se reportan métricas a dos niveles: detección binaria a nivel de oración (accuracy, precision, recall y F0.5, donde positivo indica que la frase necesita corrección) y calidad de corrección a nivel de token (GLEU, F0.5, precision, recall y distancia de edición normalizada). El modelo se evalúa sobre dos conjuntos: test sintético (distribución similar al entrenamiento) y noticias reales (evaluación de generalización fuera de dominio)

# **MBART**

```python
# ============================================================
# PIPELINE GEC — mBART-large-50
# ============================================================
#
# IMPORTANTE: ejecutar en sesión limpia, sin Unsloth importado.
# ============================================================

import sys
if "unsloth" in sys.modules:
    raise RuntimeError(
        "Unsloth está importado en esta sesión. "
        "Reinicia el kernel y ejecuta este pipeline en una sesión limpia."
    )

import json, time, gc, random, re, math
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    MBartForConditionalGeneration,
    MBart50TokenizerFast,
    get_linear_schedule_with_warmup,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN
# ============================================================

SEED       = 42
MODEL_NAME = "facebook/mbart-large-50"
SRC_LANG   = "es_XX"
TGT_LANG   = "es_XX"

# MAX_LEN=256: diagnóstico confirmó max=80 tokens con tokenizador
# mBART. Con 128 habría margen en input, pero max_length en
# generate incluye el forced_bos_token (es_XX), reduciendo el
# espacio real de generación. 256 da margen amplio sin coste.
MAX_LEN = 256

# BATCH_SIZE=2, GRAD_ACCUM=8 → batch efectivo=16.
# mBART-large-50 tiene 1.3B parámetros. Con bfloat16 ocupa ~5.5GB.
# BATCH_SIZE=2 es el máximo seguro en L4 (23.6GB) con margen para
# activaciones y gradientes.
BATCH_SIZE = 2
GRAD_ACCUM = 8

# LR=2e-5: con 5e-4 la val_loss explotó de 0.69 a 29.35 en epoch 2
# (inestabilidad numérica confirmada empíricamente). 2e-5 es el LR
# estándar para fine-tuning de mBART (Lewis et al., 2020).
LR           = 2e-5
WEIGHT_DECAY = 0.01

# N_EPOCHS=20, PATIENCE=5: con LR=2e-5 y 336 muestras mBART
# converge más lento que mT5. Con PATIENCE=3 hubo early stopping
# prematuro en epoch 4.
N_EPOCHS  = 20
PATIENCE  = 5

WARMUP_RATIO = 0.1

# NUM_BEAMS=4: beam search estándar para seq2seq preentrenados.
# A diferencia del BiLSTM (que usa argmax con TF por limitación
# arquitectural), mBART genera de forma autoregresiva real, lo
# que hace que sus métricas de detección binaria sean directamente
# válidas sin ninguna advertencia adicional.
NUM_BEAMS = 4

MAX_GRAD_NORM   = 1.0
BEST_MODEL_PATH = "best_model_mbart.pt"

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. BITSANDBYTES
# ============================================================

try:
    import bitsandbytes as bnb
    print("bitsandbytes disponible.")
except ImportError:
    import subprocess
    print("Instalando bitsandbytes...")
    subprocess.run(["pip", "install", "-q", "bitsandbytes"], check=True)
    import bitsandbytes as bnb
    print("bitsandbytes instalado.")

# ============================================================
# 3. TOKENIZADOR Y MODELO
# ============================================================

print("=" * 60)
print("CARGANDO mBART-large-50")
print("=" * 60)

clear_cache()

tokenizer          = MBart50TokenizerFast.from_pretrained(MODEL_NAME)
tokenizer.src_lang = SRC_LANG
tokenizer.tgt_lang = TGT_LANG

model = MBartForConditionalGeneration.from_pretrained(MODEL_NAME)
model = model.to(device)
model.config.use_cache = False

# forced_bos_token_id: mBART es un modelo de traducción multilingüe.
# El decoder necesita saber en qué idioma generar. Sin este parámetro
# mBART genera tokens aleatorios del vocabulario de 250k entradas.
model.generation_config.forced_bos_token_id = tokenizer.lang_code_to_id[TGT_LANG]

USE_BF16  = bool(torch.cuda.is_available() and torch.cuda.is_bf16_supported())
AMP_DTYPE = torch.bfloat16 if USE_BF16 else torch.float16
print(f"AMP dtype: {AMP_DTYPE}")

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parámetros entrenables: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 4. DIAGNÓSTICO DE LONGITUD DE TOKENS
# ============================================================

def diagnostico_tokens_mbart(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        ids = tokenizer(str(row["input"]), add_special_tokens=True)["input_ids"]
        n   = len(ids)
        if include_target:
            tgt_ids = tokenizer(
                text_target=str(row["target"]), add_special_tokens=True
            )["input_ids"]
            n = max(n, len(tgt_ids))
        lengths.append(n)
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_LEN).sum()
    print(f"\n--- Tokens mBART: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_LEN={MAX_LEN}: {truncated}/{len(lengths)}", end="")
    if truncated > 0:
        print(f"  ⚠ Sube MAX_LEN a al menos {int(lengths.max()) + 16}")
    else:
        print("  ✓")
    return lengths

print("\n" + "=" * 60)
print("DIAGNÓSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens_mbart(df_train,          "TRAIN    (max input/target)", include_target=True)
diagnostico_tokens_mbart(df_val,            "VAL      (max input/target)", include_target=True)
diagnostico_tokens_mbart(df_test,           "TEST     (solo input)",       include_target=False)
diagnostico_tokens_mbart(df_noticias_final, "NOTICIAS (solo input)",       include_target=False)

# ============================================================
# 5. DATASET Y DATALOADERS
# ============================================================

class GECDatasetMBART(Dataset):
    def __init__(self, df, tokenizer, max_len=256):
        self.src_texts = df["input"].tolist()
        self.tgt_texts = df["target"].tolist()
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.src_texts)

    def __getitem__(self, idx):
        src_enc = self.tokenizer(
            self.src_texts[idx],
            max_length=self.max_len, truncation=True,
            padding=False, return_tensors=None,
        )
        tgt_enc = self.tokenizer(
            text_target=self.tgt_texts[idx],
            max_length=self.max_len, truncation=True,
            padding=False, return_tensors=None,
        )
        return {
            "input_ids":      torch.tensor(src_enc["input_ids"],      dtype=torch.long),
            "attention_mask": torch.tensor(src_enc["attention_mask"], dtype=torch.long),
            "labels":         torch.tensor(tgt_enc["input_ids"],      dtype=torch.long),
        }

def collate_fn_mbart(batch):
    pad_id  = tokenizer.pad_token_id
    max_src = max(b["input_ids"].size(0) for b in batch)
    max_tgt = max(b["labels"].size(0)    for b in batch)

    input_ids_p = torch.full((len(batch), max_src), pad_id, dtype=torch.long)
    attn_mask_p = torch.zeros((len(batch), max_src),         dtype=torch.long)
    labels_p    = torch.full((len(batch), max_tgt), -100,   dtype=torch.long)

    for i, b in enumerate(batch):
        sl = b["input_ids"].size(0)
        tl = b["labels"].size(0)
        input_ids_p[i, :sl] = b["input_ids"]
        attn_mask_p[i, :sl] = b["attention_mask"]
        labels_p[i,    :tl] = b["labels"]

    return {
        "input_ids":      input_ids_p,
        "attention_mask": attn_mask_p,
        "labels":         labels_p,
    }

train_loader = DataLoader(
    GECDatasetMBART(df_train,          tokenizer, MAX_LEN),
    batch_size=BATCH_SIZE, shuffle=True,  collate_fn=collate_fn_mbart, pin_memory=True,
)
val_loader = DataLoader(
    GECDatasetMBART(df_val,            tokenizer, MAX_LEN),
    batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn_mbart, pin_memory=True,
)
test_loader = DataLoader(
    GECDatasetMBART(df_test,           tokenizer, MAX_LEN),
    batch_size=1, shuffle=False, collate_fn=collate_fn_mbart,
)
noticias_loader = DataLoader(
    GECDatasetMBART(df_noticias_final, tokenizer, MAX_LEN),
    batch_size=1, shuffle=False, collate_fn=collate_fn_mbart,
)

print(f"\nTrain: {len(train_loader)} batches | Val: {len(val_loader)} batches")
print(f"Test: {len(test_loader)} | Noticias: {len(noticias_loader)}")
print(f"Batch efectivo: {BATCH_SIZE} × {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ============================================================
# 6. OPTIMIZADOR Y SCHEDULER
#
# AdamW8bit sin GradScaler: GradScaler espera hooks de inf-check
# que AdamW8bit no registra. Se usa autocast solo en forward para
# reducir memoria de activaciones; backward fluye en fp32.
# ============================================================

steps_per_epoch   = math.ceil(len(train_loader) / GRAD_ACCUM)
total_optim_steps = steps_per_epoch * N_EPOCHS
warmup_steps      = int(total_optim_steps * WARMUP_RATIO)

optimizer = bnb.optim.AdamW8bit(
    model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY,
)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_optim_steps,
)

print(f"Pasos totales: {total_optim_steps} | Warmup: {warmup_steps}")
if device == "cuda":
    print(f"Memoria tras optimizador: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 7. MÉTRICAS GEC
#
# Sistema de métricas idéntico al del BiLSTM para garantizar
# comparabilidad directa en la tabla de resultados del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#
# Nota sobre validez: a diferencia del BiLSTM (que usa teacher
# forcing en evaluación), mBART genera mediante beam search
# autoregresivo real sin ver el target. Por tanto las métricas
# de detección binaria son directamente válidas y no requieren
# ninguna advertencia metodológica adicional.
# ============================================================

def decode_clean(token_ids):
    if isinstance(token_ids, torch.Tensor):
        token_ids = token_ids.cpu().tolist()
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, loss, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Loss:              {loss:.4f}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 8. FUNCIONES DE FORWARD Y EVALUACIÓN
# ============================================================

def forward_mbart(input_ids, attention_mask, labels):
    if USE_BF16:
        with torch.amp.autocast("cuda", dtype=torch.bfloat16):
            return model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels,
            )
    return model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        labels=labels,
    )


def val_loss_only(model, loader):
    """Loss con teacher forcing para early stopping. Igual que BiLSTM."""
    model.eval()
    total_loss = 0.0
    n_batches  = 0
    with torch.no_grad():
        for batch in loader:
            outputs = forward_mbart(
                batch["input_ids"].to(device),
                batch["attention_mask"].to(device),
                batch["labels"].to(device),
            )
            if not torch.isfinite(outputs.loss):
                continue
            total_loss += outputs.loss.item()
            n_batches  += 1
    return total_loss / max(n_batches, 1)


def evaluate_final(model, loader, src_texts_list=None):
    """
    Evaluación completa con beam search autoregresivo.
    Loss calculado con teacher forcing (coherente con entrenamiento).
    Métricas calculadas con generación real (beam search, sin ver tgt).
    """
    model.eval()
    total_loss       = 0.0
    n_batches        = 0
    all_metrics      = []
    predictions      = []
    decoder_start_id = tokenizer.lang_code_to_id[TGT_LANG]

    with torch.no_grad():
        for i, batch in enumerate(loader):
            input_ids      = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels         = batch["labels"].to(device)

            # loss con teacher forcing
            outputs = forward_mbart(input_ids, attention_mask, labels)
            if torch.isfinite(outputs.loss):
                total_loss += outputs.loss.item()
                n_batches  += 1

            # generación autoregresiva real con beam search
            generated = model.generate(
                input_ids=input_ids,
                attention_mask=attention_mask,
                decoder_start_token_id=decoder_start_id,
                forced_bos_token_id=decoder_start_id,
                num_beams=NUM_BEAMS,
                max_length=MAX_LEN,
                length_penalty=1.0,
                early_stopping=True,
            )

            for b in range(input_ids.size(0)):
                pred_text = decode_clean(generated[b])
                lbl_ids   = [t for t in labels[b].cpu().tolist() if t != -100]
                tgt_text  = decode_clean(lbl_ids)
                src_text  = decode_clean(input_ids[b].cpu().tolist())

                if src_texts_list is not None:
                    idx = i * loader.batch_size + b
                    if idx < len(src_texts_list):
                        src_text = src_texts_list[idx]

                metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
                all_metrics.append(metrics)
                predictions.append({
                    "src": src_text, "tgt": tgt_text, "pred": pred_text,
                    **metrics,
                })

    return total_loss / max(n_batches, 1), aggregate_metrics(all_metrics), predictions

# ============================================================
# 9. ENTRENAMIENTO
# ============================================================

print("\n" + "=" * 60)
print("ENTRENAMIENTO mBART-large-50")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

set_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.synchronize()
TRAIN_START = time.perf_counter()

best_val_loss     = float("inf")
epochs_no_improve = 0
epoch_log         = []
global_step       = 0

for epoch in range(1, N_EPOCHS + 1):
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_start = time.perf_counter()

    model.train()
    train_loss       = 0.0
    train_batches_ok = 0
    optimizer.zero_grad(set_to_none=True)

    for step, batch in enumerate(train_loader):
        input_ids      = batch["input_ids"].to(device,      non_blocking=True)
        attention_mask = batch["attention_mask"].to(device,  non_blocking=True)
        labels         = batch["labels"].to(device,          non_blocking=True)

        outputs   = forward_mbart(input_ids, attention_mask, labels)
        loss_full = outputs.loss

        if not torch.isfinite(loss_full):
            optimizer.zero_grad(set_to_none=True)
            continue

        loss = loss_full / GRAD_ACCUM
        loss.backward()
        train_loss       += loss_full.item()
        train_batches_ok += 1

        if (step + 1) % GRAD_ACCUM == 0 or (step + 1) == len(train_loader):
            torch.nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)
            global_step += 1

    train_loss = train_loss / max(train_batches_ok, 1)
    val_loss   = val_loss_only(model, val_loader)

    if val_loss < best_val_loss:
        best_val_loss     = val_loss
        epochs_no_improve = 0
        torch.save({
            "epoch":            epoch,
            "model_state_dict": model.state_dict(),
            "val_loss":         val_loss,
            "seed":             SEED,
        }, BEST_MODEL_PATH)
    else:
        epochs_no_improve += 1

    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_time = time.perf_counter() - epoch_start
    current_lr = optimizer.param_groups[0]["lr"]

    epoch_log.append({
        "epoch":        epoch,
        "train_loss":   train_loss,
        "val_loss":     val_loss,
        "lr":           current_lr,
        "epoch_time_s": epoch_time,
    })

    print(
        f"Epoch {epoch:2d}/{N_EPOCHS} | "
        f"Train: {train_loss:.4f} | Val: {val_loss:.4f} | "
        f"LR: {current_lr:.2e} | Time: {epoch_time:.2f}s | "
        f"Mem: {torch.cuda.memory_allocated(0)/1e9:.1f}GB"
    )

    if epochs_no_improve >= PATIENCE:
        print(f"\nEarly stopping en epoch {epoch}.")
        break

if torch.cuda.is_available():
    torch.cuda.synchronize()
total_train_time = time.perf_counter() - TRAIN_START

print(f"\nMejor val_loss: {best_val_loss:.4f}")
print(f"Tiempo total: {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
print(f"Tiempo promedio por época: {total_train_time/len(epoch_log):.2f}s")

# ============================================================
# 10. CARGA DEL MEJOR MODELO
# ============================================================

checkpoint = torch.load(BEST_MODEL_PATH, map_location=device)
model.load_state_dict(checkpoint["model_state_dict"])
model.generation_config.forced_bos_token_id = tokenizer.lang_code_to_id[TGT_LANG]
print(f"\nMejor modelo: epoch {checkpoint['epoch']} (val_loss={checkpoint['val_loss']:.4f})")

# ============================================================
# 11. EVALUACIÓN TEST
# ============================================================

print("\n" + "=" * 60)
print("EVALUACIÓN EN TEST SET — mBART-large-50")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

EVAL_TEST_START = time.perf_counter()
test_loss, test_agg, test_preds = evaluate_final(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_loss, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 12. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "=" * 60)
print("EVALUACIÓN EN NOTICIAS — mBART-large-50")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

EVAL_NOT_START = time.perf_counter()
noticias_loss, noticias_agg, noticias_preds = evaluate_final(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_loss, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 13. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":      "mBART-large-50",
    "seed":            SEED,
    "device":          device,
    "n_params":        n_params,
    "batch_size":      BATCH_SIZE,
    "grad_accum":      GRAD_ACCUM,
    "effective_batch": BATCH_SIZE * GRAD_ACCUM,
    "max_len":         MAX_LEN,
    "lr":              LR,
    "optimizer":       "AdamW8bit",
    "mixed_precision": "bf16_autocast" if USE_BF16 else "fp32",
    "num_beams":       NUM_BEAMS,
    "decode_strategy": f"beam_search_n{NUM_BEAMS}",

    "epochs_run":         len(epoch_log),
    "best_epoch":         checkpoint["epoch"],
    "best_val_loss":      float(best_val_loss),
    "total_train_time_s": total_train_time,
    "avg_epoch_time_s":   total_train_time / len(epoch_log),

    # TEST — clasificación binaria
    "test_loss":              test_loss,
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_loss":              noticias_loss,
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES — RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mbart.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
        "epoch_log":            epoch_log,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mbart.json'")
```

**Metodología:**

Se aplica fine-tuning sobre mBART-large-50 (modelo **encoder-decoder preentrenado** en 50 idiomas incluyendo español, 610M parámetros) con el tokenizador MBart50TokenizerFast con lengua fuente y destino fijadas a es_XX. El entrenamiento emplea el optimizador AdamW8bit (lr=2e-5, weight decay=0.01) con mixed precision bfloat16, gradient accumulation de 8 pasos para un batch efectivo de 16, scheduler con warmup lineal del 10% seguido de decaimiento lineal, gradient clipping (norma 1.0) y early stopping con paciencia 5 sobre un máximo de 20 épocas (~336 muestras de entrenamiento). La generación se realiza mediante **beam search autoregresivo con 4 beams** sin acceso al target en ningún momento. Se reportan métricas a dos niveles: detección binaria a nivel de oración —donde positivo indica que la frase requiere corrección— (accuracy, precision, recall y F0.5, métrica estándar en GEC que penaliza más las correcciones erróneas que los errores no detectados) y calidad de corrección a nivel de token (GLEU, F0.5 tok, precision, recall y distancia de edición normalizada). El modelo se evalúa sobre test sintético (misma distribución que el entrenamiento) y noticias reales (generalización fuera de dominio).

# **MT5**

```python
# ============================================================
# PIPELINE GEC — mT5-base
# ============================================================

import json, time, gc, random, re, math
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    MT5ForConditionalGeneration,
    T5TokenizerFast,
    get_linear_schedule_with_warmup,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN
#
# TASK_PREFIX="correct: ": prefijo estándar para modelos T5/mT5
#   en tareas de corrección de texto. Indica al modelo la tarea
#   sin necesidad de arquitectura específica.
#
# MAX_LEN=256: diagnóstico confirmó max=92 tokens con prefijo.
#   256 da margen amplio sin coste de memoria.
#
# BATCH_SIZE=4, GRAD_ACCUM=4 → batch efectivo=16.
#   mT5-base ocupa ~3.9GB en memoria. Con batch=4 y bfloat16
#   el consumo total es ~5.7GB, seguro en L4 (23.6GB).
#   Batch físico mayor que mBART porque mT5-base es más ligero.
#
# LR=5e-4: mT5 necesita LR más alto que mBART. Con 2e-5 la
#   convergencia era demasiado lenta (confirmado empíricamente).
#   5e-4 es el LR recomendado para fine-tuning de mT5 en tareas
#   de generación de texto (Xue et al., 2021).
#
# N_EPOCHS=15, PATIENCE=5: con LR=5e-4 mT5 converge rápido;
#   en los experimentos el óptimo fue epoch 8 de 10. Se sube
#   a 15 épocas para dar margen por si en otra ejecución
#   converge más tarde, sin riesgo de overfitting porque
#   PATIENCE=5 detiene el entrenamiento si no mejora.
#   PATIENCE=5 es consistente con mBART.
#
# NUM_BEAMS=4: beam search estándar para seq2seq preentrenados.
# ============================================================

SEED            = 42
MODEL_NAME      = "google/mt5-base"
TASK_PREFIX     = "correct: "
MAX_LEN         = 256
BATCH_SIZE      = 4
GRAD_ACCUM      = 4
LR              = 5e-4
WEIGHT_DECAY    = 0.01
N_EPOCHS        = 15
PATIENCE        = 5
WARMUP_RATIO    = 0.1
NUM_BEAMS       = 4
MAX_GRAD_NORM   = 1.0
BEST_MODEL_PATH = "best_model_mt5.pt"

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. BITSANDBYTES
# ============================================================

try:
    import bitsandbytes as bnb
    print("bitsandbytes disponible.")
except ImportError:
    import subprocess
    print("Instalando bitsandbytes...")
    subprocess.run(["pip", "install", "-q", "bitsandbytes"], check=True)
    import bitsandbytes as bnb
    print("bitsandbytes instalado.")

# ============================================================
# 3. TOKENIZADOR Y MODELO
# ============================================================

print("=" * 60)
print("CARGANDO mT5-base")
print("=" * 60)

clear_cache()

tokenizer = T5TokenizerFast.from_pretrained(MODEL_NAME)
model     = MT5ForConditionalGeneration.from_pretrained(MODEL_NAME)
model     = model.to(device)
model.config.use_cache = False

USE_BF16  = bool(torch.cuda.is_available() and torch.cuda.is_bf16_supported())
AMP_DTYPE = torch.bfloat16 if USE_BF16 else torch.float32
print(f"AMP dtype: {AMP_DTYPE}")

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parámetros entrenables: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 4. DIAGNÓSTICO DE LONGITUD DE TOKENS
# ============================================================

def diagnostico_tokens_mt5(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        src = TASK_PREFIX + str(row["input"])
        ids = tokenizer(src, add_special_tokens=True)["input_ids"]
        n   = len(ids)
        if include_target:
            tgt_ids = tokenizer(
                text_target=str(row["target"]), add_special_tokens=True
            )["input_ids"]
            n = max(n, len(tgt_ids))
        lengths.append(n)
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_LEN).sum()
    print(f"\n--- Tokens mT5: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_LEN={MAX_LEN}: {truncated}/{len(lengths)}", end="")
    if truncated > 0:
        print(f"  ⚠ Sube MAX_LEN a al menos {int(lengths.max()) + 16}")
    else:
        print("  ✓")
    return lengths

print("\n" + "=" * 60)
print("DIAGNÓSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens_mt5(df_train,          "TRAIN    (max input/target)", include_target=True)
diagnostico_tokens_mt5(df_val,            "VAL      (max input/target)", include_target=True)
diagnostico_tokens_mt5(df_test,           "TEST     (solo input)",       include_target=False)
diagnostico_tokens_mt5(df_noticias_final, "NOTICIAS (solo input)",       include_target=False)

# ============================================================
# 5. DATASET Y DATALOADERS
# ============================================================

class GECDatasetMT5(Dataset):
    def __init__(self, df, tokenizer, max_len=256, prefix="correct: "):
        self.raw_src_texts = df["input"].astype(str).tolist()
        self.src_texts     = [prefix + t for t in self.raw_src_texts]
        self.tgt_texts     = df["target"].astype(str).tolist()
        self.tokenizer     = tokenizer
        self.max_len       = max_len

    def __len__(self):
        return len(self.src_texts)

    def __getitem__(self, idx):
        src_enc = self.tokenizer(
            self.src_texts[idx],
            max_length=self.max_len,
            truncation=True,
            padding=False,
            return_tensors=None,
        )
        tgt_enc = self.tokenizer(
            text_target=self.tgt_texts[idx],
            max_length=self.max_len,
            truncation=True,
            padding=False,
            return_tensors=None,
        )
        return {
            "input_ids":      torch.tensor(src_enc["input_ids"],      dtype=torch.long),
            "attention_mask": torch.tensor(src_enc["attention_mask"], dtype=torch.long),
            "labels":         torch.tensor(tgt_enc["input_ids"],      dtype=torch.long),
        }

def collate_fn_mt5(batch):
    pad_id  = tokenizer.pad_token_id
    max_src = max(b["input_ids"].size(0) for b in batch)
    max_tgt = max(b["labels"].size(0)    for b in batch)

    input_ids_p = torch.full((len(batch), max_src), pad_id, dtype=torch.long)
    attn_mask_p = torch.zeros((len(batch), max_src),         dtype=torch.long)
    labels_p    = torch.full((len(batch), max_tgt), -100,   dtype=torch.long)

    for i, b in enumerate(batch):
        sl = b["input_ids"].size(0)
        tl = b["labels"].size(0)
        input_ids_p[i, :sl] = b["input_ids"]
        attn_mask_p[i, :sl] = b["attention_mask"]
        labels_p[i,    :tl] = b["labels"]

    return {
        "input_ids":      input_ids_p,
        "attention_mask": attn_mask_p,
        "labels":         labels_p,
    }

train_loader = DataLoader(
    GECDatasetMT5(df_train,          tokenizer, MAX_LEN, TASK_PREFIX),
    batch_size=BATCH_SIZE, shuffle=True,  collate_fn=collate_fn_mt5, pin_memory=True,
)
val_loader = DataLoader(
    GECDatasetMT5(df_val,            tokenizer, MAX_LEN, TASK_PREFIX),
    batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn_mt5, pin_memory=True,
)
test_loader = DataLoader(
    GECDatasetMT5(df_test,           tokenizer, MAX_LEN, TASK_PREFIX),
    batch_size=1, shuffle=False, collate_fn=collate_fn_mt5,
)
noticias_loader = DataLoader(
    GECDatasetMT5(df_noticias_final, tokenizer, MAX_LEN, TASK_PREFIX),
    batch_size=1, shuffle=False, collate_fn=collate_fn_mt5,
)

print(f"\nTrain: {len(train_loader)} batches | Val: {len(val_loader)} batches")
print(f"Test: {len(test_loader)} | Noticias: {len(noticias_loader)}")
print(f"Batch efectivo: {BATCH_SIZE} × {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ============================================================
# 6. OPTIMIZADOR Y SCHEDULER
# ============================================================

steps_per_epoch   = math.ceil(len(train_loader) / GRAD_ACCUM)
total_optim_steps = steps_per_epoch * N_EPOCHS
warmup_steps      = int(total_optim_steps * WARMUP_RATIO)

optimizer = bnb.optim.AdamW8bit(
    model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY,
)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_optim_steps,
)

print(f"Pasos totales: {total_optim_steps} | Warmup: {warmup_steps}")
if device == "cuda":
    print(f"Memoria tras optimizador: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 7. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM y mBART para
# garantizar comparabilidad directa en la tabla del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#
# clean_extra_ids: mT5 puede generar tokens <extra_id_N> en
#   condiciones de baja confianza. Se eliminan en postprocesado
#   para evitar que contaminen las métricas de calidad.
# ============================================================

def clean_extra_ids(text):
    text = re.sub(r"<extra_id_\d+>", "", text)
    return re.sub(r"\s+", " ", text).strip()

def decode_clean(token_ids):
    if isinstance(token_ids, torch.Tensor):
        token_ids = token_ids.cpu().tolist()
    text = tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()
    return clean_extra_ids(text)


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, loss, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Loss:              {loss:.4f}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 8. FUNCIONES DE FORWARD Y EVALUACIÓN
# ============================================================

def forward_mt5(input_ids, attention_mask, labels):
    if USE_BF16:
        with torch.amp.autocast("cuda", dtype=torch.bfloat16):
            return model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels,
            )
    return model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        labels=labels,
    )


def val_loss_only(model, loader):
    """Loss con teacher forcing para early stopping."""
    model.eval()
    total_loss = 0.0
    n_batches  = 0
    with torch.no_grad():
        for batch in loader:
            outputs = forward_mt5(
                batch["input_ids"].to(device),
                batch["attention_mask"].to(device),
                batch["labels"].to(device),
            )
            if not torch.isfinite(outputs.loss):
                continue
            total_loss += outputs.loss.item()
            n_batches  += 1
    return total_loss / max(n_batches, 1)


def evaluate_final(model, loader, src_texts_list=None):
    """
    Loss con teacher forcing (coherente con entrenamiento) +
    métricas con beam search autoregresivo (predicción honesta).
    """
    model.eval()
    total_loss  = 0.0
    n_batches   = 0
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, batch in enumerate(loader):
            input_ids      = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels         = batch["labels"].to(device)

            # loss con teacher forcing
            outputs = forward_mt5(input_ids, attention_mask, labels)
            if torch.isfinite(outputs.loss):
                total_loss += outputs.loss.item()
                n_batches  += 1

            # generación autoregresiva real con beam search
            generated = model.generate(
                input_ids=input_ids,
                attention_mask=attention_mask,
                num_beams=NUM_BEAMS,
                max_length=MAX_LEN,
                length_penalty=1.0,
                early_stopping=True,
            )

            for b in range(input_ids.size(0)):
                pred_text = decode_clean(generated[b])
                lbl_ids   = [t for t in labels[b].cpu().tolist() if t != -100]
                tgt_text  = decode_clean(lbl_ids)

                # eliminar el prefijo "correct: " del src decodificado
                src_raw  = decode_clean(input_ids[b].cpu().tolist())
                src_text = src_raw.replace(TASK_PREFIX, "", 1).strip()

                # preferir texto original del dataframe (más limpio)
                if src_texts_list is not None:
                    idx = i * loader.batch_size + b
                    if idx < len(src_texts_list):
                        src_text = src_texts_list[idx]

                metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
                all_metrics.append(metrics)
                predictions.append({
                    "src": src_text, "tgt": tgt_text, "pred": pred_text,
                    **metrics,
                })

    return total_loss / max(n_batches, 1), aggregate_metrics(all_metrics), predictions

# ============================================================
# 9. ENTRENAMIENTO
# ============================================================

print("\n" + "=" * 60)
print("ENTRENAMIENTO mT5-base")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

set_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.synchronize()
TRAIN_START = time.perf_counter()

best_val_loss     = float("inf")
epochs_no_improve = 0
epoch_log         = []
global_step       = 0

for epoch in range(1, N_EPOCHS + 1):
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_start = time.perf_counter()

    model.train()
    train_loss       = 0.0
    train_batches_ok = 0
    optimizer.zero_grad(set_to_none=True)

    for step, batch in enumerate(train_loader):
        input_ids      = batch["input_ids"].to(device,      non_blocking=True)
        attention_mask = batch["attention_mask"].to(device,  non_blocking=True)
        labels         = batch["labels"].to(device,          non_blocking=True)

        outputs   = forward_mt5(input_ids, attention_mask, labels)
        loss_full = outputs.loss

        if not torch.isfinite(loss_full):
            optimizer.zero_grad(set_to_none=True)
            continue

        loss = loss_full / GRAD_ACCUM
        loss.backward()
        train_loss       += loss_full.item()
        train_batches_ok += 1

        if (step + 1) % GRAD_ACCUM == 0 or (step + 1) == len(train_loader):
            torch.nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)
            global_step += 1

    train_loss = train_loss / max(train_batches_ok, 1)
    val_loss   = val_loss_only(model, val_loader)

    if val_loss < best_val_loss:
        best_val_loss     = val_loss
        epochs_no_improve = 0
        torch.save({
            "epoch":            epoch,
            "model_state_dict": model.state_dict(),
            "val_loss":         val_loss,
            "seed":             SEED,
        }, BEST_MODEL_PATH)
    else:
        epochs_no_improve += 1

    if torch.cuda.is_available():
        torch.cuda.synchronize()
    epoch_time = time.perf_counter() - epoch_start
    current_lr = optimizer.param_groups[0]["lr"]

    epoch_log.append({
        "epoch":        epoch,
        "train_loss":   train_loss,
        "val_loss":     val_loss,
        "lr":           current_lr,
        "epoch_time_s": epoch_time,
    })

    print(
        f"Epoch {epoch:2d}/{N_EPOCHS} | "
        f"Train: {train_loss:.4f} | Val: {val_loss:.4f} | "
        f"LR: {current_lr:.2e} | Time: {epoch_time:.2f}s | "
        f"Mem: {torch.cuda.memory_allocated(0)/1e9:.1f}GB"
    )

    if epochs_no_improve >= PATIENCE:
        print(f"\nEarly stopping en epoch {epoch}.")
        break

if torch.cuda.is_available():
    torch.cuda.synchronize()
total_train_time = time.perf_counter() - TRAIN_START

print(f"\nMejor val_loss: {best_val_loss:.4f}")
print(f"Tiempo total: {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
print(f"Tiempo promedio por época: {total_train_time/len(epoch_log):.2f}s")

# ============================================================
# 10. CARGA DEL MEJOR MODELO
# ============================================================

checkpoint = torch.load(BEST_MODEL_PATH, map_location=device)
model.load_state_dict(checkpoint["model_state_dict"])
print(f"\nMejor modelo: epoch {checkpoint['epoch']} (val_loss={checkpoint['val_loss']:.4f})")

# ============================================================
# 11. EVALUACIÓN TEST
# ============================================================

print("\n" + "=" * 60)
print("EVALUACIÓN EN TEST SET — mT5-base")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

EVAL_TEST_START = time.perf_counter()
test_loss, test_agg, test_preds = evaluate_final(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_loss, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 12. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "=" * 60)
print("EVALUACIÓN EN NOTICIAS — mT5-base")
print("=" * 60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

EVAL_NOT_START = time.perf_counter()
noticias_loss, noticias_agg, noticias_preds = evaluate_final(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_loss, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 13. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":      "mT5-base",
    "seed":            SEED,
    "device":          device,
    "n_params":        n_params,
    "task_prefix":     TASK_PREFIX,
    "batch_size":      BATCH_SIZE,
    "grad_accum":      GRAD_ACCUM,
    "effective_batch": BATCH_SIZE * GRAD_ACCUM,
    "max_len":         MAX_LEN,
    "lr":              LR,
    "optimizer":       "AdamW8bit",
    "mixed_precision": "bf16_autocast" if USE_BF16 else "fp32",
    "num_beams":       NUM_BEAMS,
    "decode_strategy": f"beam_search_n{NUM_BEAMS}",

    "epochs_run":         len(epoch_log),
    "best_epoch":         checkpoint["epoch"],
    "best_val_loss":      float(best_val_loss),
    "total_train_time_s": total_train_time,
    "avg_epoch_time_s":   total_train_time / len(epoch_log),

    # TEST — clasificación binaria
    "test_loss":              test_loss,
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_loss":              noticias_loss,
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES — RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mt5.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
        "epoch_log":            epoch_log,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mt5.json'")
```

**Metodología:**

Se aplica fine-tuning sobre mT5-base (modelo **encoder-decoder preentrenado** en 101 idiomas incluyendo español, 580M parámetros) adaptándolo a la tarea de GEC mediante el tokenizador T5TokenizerFast con el prefijo de tarea "correct: ". El entrenamiento emplea el optimizador AdamW8bit (lr=5e-4, weight decay=0.01) con mixed precision bfloat16, gradient accumulation de 4 pasos para un batch efectivo de 16, scheduler con warmup lineal del 10% seguido de decaimiento lineal, gradient clipping (norma 1.0) y early stopping con paciencia 5 sobre un máximo de 15 épocas (~336 muestras de entrenamiento). El learning rate de 5e-4 es significativamente más alto que el empleado para mBART, lo cual es consistente con la literatura de fine-tuning de modelos T5 (Xue et al., 2021), donde valores en torno a 1e-3 son habituales al entrenar desde preentrenamiento multilingual sin adaptación previa a la tarea. La generación se realiza mediante **beam search autoregresivo con 4 beams** sin acceso al target. Se reportan métricas a dos niveles: detección binaria a nivel de oración (accuracy, precision, recall y F0.5) y calidad de corrección a nivel de token (GLEU, F0.5 tok, precision, recall y distancia de edición normalizada). El modelo se evalúa sobre test sintético y noticias reales.

*Referencias:*
Xue, L., Constant, N., Roberts, A., Kale, M., Al-Rfou, R., Siddhant, A., Barua, A., & Raffel, C. (2021). mT5: A massively multilingual pre-trained text-to-text transformer. In Proceedings of the 2021 Conference of the North American Chapter of the Association for Computational Linguistics: Human Language Technologies (NAACL-HLT 2021), pp. 483–498. Association for Computational Linguistics.

**Llama 3.2 3B Instruct: Prompt 1 (zero-shot)**


```python
# ============================================================
# PIPELINE GEC — Llama 3.2 3B Instruct (zero-shot)
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 3B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#
# max_new_tokens=150: límite generoso para frases largas.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~60 tokens
#   de Llama. 150 deja margen sin truncar.
#
# repetition_penalty=1.0: sin penalización adicional.
#   Se mantiene igual que en el setup usado para tú impersonal
#   para preservar la comparabilidad experimental entre fenómenos.
#
# COMPARABILIDAD:
#   Llama se evalúa zero-shot (sin fine-tuning).
#   BiLSTM, mBART y mT5 se evaluaron con fine-tuning.
#   Las métricas de detección binaria y calidad token-level
#   se calculan con la misma lógica que el resto de modelos
#   para mantener comparabilidad en la tabla del paper.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "meta-llama/Llama-3.2-3B-Instruct"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Llama 3.2 3B de ~6GB (bf16) a ~1.8GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.4GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# device_map="auto": distribuye capas automáticamente en GPU.
# ============================================================

print("="*60)
print("CARGANDO LLAMA 3.2 3B EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Cuando el gerundio exprese una acción posterior a la del verbo principal, sustitúyelo por una construcción con 'y' y verbo conjugado. "
    "No modifiques nada más. "
    "Devuelve solo la oración final. "
    "Si no es necesario corregir, devuelve la oración exactamente igual."
)

USER_PROMPT = "{sentence}"

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetLlama(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetLlama(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetLlama(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART y mT5
# para garantizar comparabilidad directa en la tabla del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Llama.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Llama:
    - Elimina comillas, guiones iniciales, "Corrección:" etc.
    - Llama Instruct a veces añade estas marcas incluso con
      la instrucción explícita de no hacerlo.
    """
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = re.sub(r'^(corrección|corregido|texto corregido)\s*:\s*',
                  '', text, flags=re.IGNORECASE)
    text = text.strip().strip('"\'')
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_llama(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Llama 3.2 3B (zero-shot)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_llama(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Llama 3.2 3B (zero-shot)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_llama(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Llama-3.2-3B-Instruct",
    "evaluation_mode":   "zero_shot",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama.json'")
```

**Metodología:**

Se aplica inferencia zero-shot sobre **Llama 3.2 3B Instruct** (modelo **decoder-only causal**, 3B parámetros) cuantizado en 4 bits con formato NF4 (double quantization, compute dtype bfloat16) mediante BitsAndBytesConfig, lo que permite ejecutar el modelo con recursos reducidos sin reentrenamiento. La generación se realiza con decodificación greedy (do_sample=False, num_beams=1, repetition_penalty=1.0, max_new_tokens=150) y batch_size=1, obligatorio para LLMs causales en inferencia secuencial. Con el objetivo de analizar el efecto de la estrategia de prompting sobre el rendimiento del modelo, se evalúan cuatro configuraciones de prompt manteniendo fijos el modelo, los hiperparámetros de generación y el conjunto de evaluación: (1) zero-shot, (2) zero-shot estructurado, (3) few-shot y (4) few-shot optimizado. Se reportan métricas a dos niveles: detección binaria a nivel de oración (accuracy, precision, recall y F0.5) y calidad de corrección a nivel de token (GLEU, F0.5 tok, precision, recall y distancia de edición normalizada). El modelo se evalúa sobre test sintético y noticias reales.

**Llama 3.2 3B Instruct: NIVEL 2 — Zero-shot guiado con reglas claras**

```python
# ============================================================
# PIPELINE GEC — Llama 3.2 3B Instruct (zero-shot)
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 3B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#
# max_new_tokens=150: límite generoso para frases largas.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~60 tokens
#   de Llama. 150 deja margen sin truncar.
#
# repetition_penalty=1.0: sin penalización adicional.
#   Se mantiene igual que en el setup usado para tú impersonal
#   para preservar la comparabilidad experimental entre fenómenos.
#
# COMPARABILIDAD:
#   Llama se evalúa zero-shot (sin fine-tuning).
#   BiLSTM, mBART y mT5 se evaluaron con fine-tuning.
#   Las métricas de detección binaria y calidad token-level
#   se calculan con la misma lógica que el resto de modelos
#   para mantener comparabilidad en la tabla del paper.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "meta-llama/Llama-3.2-3B-Instruct"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Llama 3.2 3B de ~6GB (bf16) a ~1.8GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.4GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# device_map="auto": distribuye capas automáticamente en GPU.
# ============================================================

print("="*60)
print("CARGANDO LLAMA 3.2 3B EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================
SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal.\n\n"

    "TAREA:\n"
    "Detecta y corrige únicamente el gerundio de posterioridad.\n\n"

    "REGLA PRINCIPAL:\n"
    "Existe gerundio de posterioridad cuando el gerundio expresa una acción que ocurre después de la acción del verbo principal. "
    "En esos casos, debe reformularse mediante una coordinación con 'y' y un verbo conjugado.\n\n"

    "CÓMO CORREGIR:\n"
    "- Corrige solo el gerundio de posterioridad.\n"
    "- Sustituye el gerundio por 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo significado.\n"
    "- Ajusta pronombres o concordancia solo si es estrictamente necesario.\n"
    "- Modifica solo lo mínimo imprescindible.\n"
    "- No reformules la oración.\n\n"

    "NO CORREGIR EN ESTOS CASOS:\n"
    "- Si el gerundio expresa simultaneidad.\n"
    "- Si el gerundio expresa modo, medio, causa, condición o descripción.\n"
    "- Si forma parte de una perífrasis verbal o construcción aspectual (por ejemplo: 'está estudiando', 'sigue trabajando', 'terminó aceptando', 'acabó cediendo').\n"
    "- Si aparece dentro de comillas o estilo directo.\n"
    "- Si no hay gerundio.\n"
    "- Si el fenómeno no está presente.\n\n"

    "IMPORTANTE:\n"
    "- No cambies palabras innecesariamente.\n"
    "- No mejores el estilo.\n"
    "- No añadas ni elimines información.\n"
    "- Si no estás seguro de que el gerundio expresa posterioridad, no hagas cambios.\n\n"

    "SALIDA:\n"
    "Devuelve únicamente la oración final corregida. "
    "Si no es necesario corregir, devuelve la oración exactamente igual."
)

USER_PROMPT = "{sentence}"

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetLlama(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetLlama(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetLlama(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART y mT5
# para garantizar comparabilidad directa en la tabla del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Llama.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Llama:
    - Elimina comillas, guiones iniciales, "Corrección:" etc.
    - Llama Instruct a veces añade estas marcas incluso con
      la instrucción explícita de no hacerlo.
    """
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = re.sub(r'^(corrección|corregido|texto corregido)\s*:\s*',
                  '', text, flags=re.IGNORECASE)
    text = text.strip().strip('"\'')
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_llama(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Llama 3.2 3B")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_llama(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Llama 3.2 3B (NIVEL 2 — Zero-shot guiado con reglas claras)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_llama(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Llama-3.2-3B-Instruct",
    "evaluation_mode":   "NIVEL 2 — Zero-shot guiado con reglas claras",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama.json'")
```

**Llama 3.2 3B Instruct: Prompt 3 (few-shot)**

```python
# ============================================================
# PIPELINE GEC — Llama 3.2 3B Instruct
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 3B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#
# max_new_tokens=150: límite generoso para frases largas.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~60 tokens
#   de Llama. 150 deja margen sin truncar.
#
# repetition_penalty=1.0: sin penalización adicional.
#   Se mantiene igual que en el setup usado para tú impersonal
#   para preservar la comparabilidad experimental entre fenómenos.
#
# COMPARABILIDAD:
#   Llama se evalúa zero-shot (sin fine-tuning).
#   BiLSTM, mBART y mT5 se evaluaron con fine-tuning.
#   Las métricas de detección binaria y calidad token-level
#   se calculan con la misma lógica que el resto de modelos
#   para mantener comparabilidad en la tabla del paper.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "meta-llama/Llama-3.2-3B-Instruct"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Llama 3.2 3B de ~6GB (bf16) a ~1.8GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.4GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# device_map="auto": distribuye capas automáticamente en GPU.
# ============================================================

print("="*60)
print("CARGANDO LLAMA 3.2 3B EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Devuelve solo la oración final."
)

USER_PROMPT = (
    "Input: La investigadora cerró la fase de recogida de datos, redactando el informe semanas después.\n"
    "Output: La investigadora cerró la fase de recogida de datos y redactó el informe semanas después.\n\n"

    "Input: El técnico revisará el servidor principal, reiniciándolo más tarde para aplicar la configuración definitiva.\n"
    "Output: El técnico revisará el servidor principal y lo reiniciará más tarde para aplicar la configuración definitiva.\n\n"

    "Input: El museo restauró la pieza dañada, exhibiéndola meses después en una sala temporal.\n"
    "Output: El museo restauró la pieza dañada y la exhibió meses después en una sala temporal.\n\n"

    "Input: La abogada preparó la documentación, presentándola al día siguiente en el registro.\n"
    "Output: La abogada preparó la documentación y la presentó al día siguiente en el registro.\n\n"

    "Input: El alumnado siguió trabajando en silencio durante la prueba.\n"
    "Output: El alumnado siguió trabajando en silencio durante la prueba.\n\n"

    "Input: El conferenciante habló sonriendo al público durante toda la sesión.\n"
    "Output: El conferenciante habló sonriendo al público durante toda la sesión.\n\n"

    "Input: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n"
    "Output: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n\n"

    "Input: El equipo terminó aceptando la propuesta tras varias horas de debate.\n"
    "Output: El equipo terminó aceptando la propuesta tras varias horas de debate.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetLlama(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetLlama(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetLlama(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART y mT5
# para garantizar comparabilidad directa en la tabla del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Llama.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Llama:
    - Elimina comillas, guiones iniciales, "Corrección:" etc.
    - Llama Instruct a veces añade estas marcas incluso con
      la instrucción explícita de no hacerlo.
    """
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = re.sub(r'^(corrección|corregido|texto corregido)\s*:\s*',
                  '', text, flags=re.IGNORECASE)
    text = text.strip().strip('"\'')
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_llama(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Llama 3.2 3B")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_llama(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Llama 3.2 3B (NIVEL 3 — Few-shot con pocas reglas")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_llama(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Llama-3.2-3B-Instruct",
    "evaluation_mode":   "NIVEL 3 — Few-shot con pocas reglas",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama.json'")
```

**Llama 3.2 3B Instruct: Prompt 4 (NIVEL 4 — Few-shot + reglas)**

```python
# ============================================================
# PIPELINE GEC — Llama 3.2 3B Instruct
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 3B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#
# max_new_tokens=150: límite generoso para frases largas.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~60 tokens
#   de Llama. 150 deja margen sin truncar.
#
# repetition_penalty=1.0: sin penalización adicional.
#   Se mantiene igual que en el setup usado para tú impersonal
#   para preservar la comparabilidad experimental entre fenómenos.
#
# COMPARABILIDAD:
#   Llama se evalúa zero-shot (sin fine-tuning).
#   BiLSTM, mBART y mT5 se evaluaron con fine-tuning.
#   Las métricas de detección binaria y calidad token-level
#   se calculan con la misma lógica que el resto de modelos
#   para mantener comparabilidad en la tabla del paper.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "meta-llama/Llama-3.2-3B-Instruct"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Llama 3.2 3B de ~6GB (bf16) a ~1.8GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.4GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# device_map="auto": distribuye capas automáticamente en GPU.
# ============================================================

print("="*60)
print("CARGANDO LLAMA 3.2 3B EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal y en un único fenómeno: "
    "el gerundio de posterioridad.\n\n"

    "TAREA:\n"
    "Corrige SOLO los casos en los que el gerundio expresa una acción posterior a la del verbo principal.\n\n"

    "DEFINICIÓN:\n"
    "Hay gerundio de posterioridad únicamente cuando la acción expresada por el gerundio ocurre después de la acción del verbo principal. "
    "Si el gerundio expresa simultaneidad, modo, medio, causa, condición, descripción o forma parte de una perífrasis, no debe corregirse.\n\n"

    "OBJETIVO:\n"
    "Transforma esos casos en una coordinación con 'y' y un verbo conjugado.\n\n"

    "REGLA PRINCIPAL:\n"
    "Si hay duda sobre si el gerundio expresa posterioridad real, NO corrijas.\n\n"

    "CORREGIR SOLO SI:\n"
    "- el gerundio expresa con claridad una acción posterior;\n"
    "- la acción del gerundio ocurre después de la acción del verbo principal;\n"
    "- la corrección puede hacerse sustituyendo el gerundio por 'y' + verbo conjugado;\n"
    "- el nuevo verbo puede mantenerse en el mismo tiempo y modo que el verbo principal sin reformular libremente la oración.\n\n"

    "CÓMO CORREGIR:\n"
    "- Sustituye solo el segmento en gerundio afectado.\n"
    "- Usa 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo sujeto.\n"
    "- Ajusta pronombres o concordancia solo si es imprescindible para que la oración sea gramatical.\n"
    "- Conserva el significado original.\n\n"

    "NO CORREGIR NUNCA SI:\n"
    "- el gerundio expresa simultaneidad con el verbo principal;\n"
    "- el gerundio expresa modo o manera: 'habló sonriendo', 'respondió mirando al suelo';\n"
    "- el gerundio expresa medio o instrumento;\n"
    "- el gerundio expresa causa, condición o descripción;\n"
    "- el gerundio forma parte de una perífrasis o construcción aspectual: 'está estudiando', 'sigue trabajando', 'continúa avanzando', 'acabó aceptando', 'terminó cediendo', 'lleva años investigando';\n"
    "- el gerundio aparece dentro de comillas, cita directa o estilo directo;\n"
    "- la frase no contiene gerundio;\n"
    "- la frase ya es correcta y no contiene gerundio de posterioridad;\n"
    "- la modificación exigiría reescribir libremente la oración o alterar partes ajenas al fenómeno.\n\n"

    "RESTRICCIONES DE SALIDA:\n"
    "- Devuelve una sola oración final.\n"
    "- No expliques nada.\n"
    "- No añadas comentarios.\n"
    "- No cambies palabras no relacionadas con el fenómeno.\n"
    "- No cambies léxico, nombres, números, puntuación, mayúsculas, comillas ni tildes, salvo lo imprescindible para la corrección mínima.\n"
    "- No resumas ni reformules.\n"
    "- Conserva exactamente el contenido original excepto los cambios mínimos necesarios para sustituir el gerundio de posterioridad.\n\n"

    "CRITERIO DE CONSERVACIÓN:\n"
    "Si solo una parte de la oración contiene un gerundio de posterioridad, corrige solo esa parte y deja intacto todo lo demás.\n\n"

    "IMPORTANTE:\n"
    "Es mejor dejar una frase sin corregir que hacer una corrección falsa."
)

USER_PROMPT = (
    "Ejemplos:\n\n"

    "Input: El laboratorio completó la primera fase del ensayo, publicando los resultados dos meses después.\n"
    "Output: El laboratorio completó la primera fase del ensayo y publicó los resultados dos meses después.\n\n"

    "Input: La comisaria inaugurará la exposición temporal, trasladándola después a la sede principal del museo.\n"
    "Output: La comisaria inaugurará la exposición temporal y la trasladará después a la sede principal del museo.\n\n"

    "Input: El decano presentó el nuevo plan docente, remitiéndolo al claustro al día siguiente para su valoración.\n"
    "Output: El decano presentó el nuevo plan docente y lo remitió al claustro al día siguiente para su valoración.\n\n"

    "Input: La empresa validó el prototipo final, registrando la patente semanas más tarde.\n"
    "Output: La empresa validó el prototipo final y registró la patente semanas más tarde.\n\n"

    "Input: El bibliotecario ordenará los manuscritos antiguos, digitalizándolos después para el archivo central.\n"
    "Output: El bibliotecario ordenará los manuscritos antiguos y los digitalizará después para el archivo central.\n\n"

    "Input: La arquitecta restauró la fachada principal, reabriendo el edificio meses después al público general.\n"
    "Output: La arquitecta restauró la fachada principal y reabrió el edificio meses después al público general.\n\n"

    "Input: El rector compareció ante la prensa, anunciando la decisión definitiva horas más tarde en un comunicado oficial.\n"
    "Output: El rector compareció ante la prensa y anunció la decisión definitiva horas más tarde en un comunicado oficial.\n\n"

    "Input: El equipo entregó la versión preliminar del informe, incorporando las correcciones la semana siguiente.\n"
    "Output: El equipo entregó la versión preliminar del informe e incorporó las correcciones la semana siguiente.\n\n"

    "Input: La profesora explicó el tema utilizando ejemplos del examen anterior.\n"
    "Output: La profesora explicó el tema utilizando ejemplos del examen anterior.\n\n"

    "Input: El violinista interpretó la pieza mirando al director en los compases finales.\n"
    "Output: El violinista interpretó la pieza mirando al director en los compases finales.\n\n"

    "Input: El departamento sigue revisando las solicitudes pendientes.\n"
    "Output: El departamento sigue revisando las solicitudes pendientes.\n\n"

    "Input: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n"
    "Output: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n\n"

    "Input: \"Estoy preparando la defensa\", respondió la doctoranda.\n"
    "Output: \"Estoy preparando la defensa\", respondió la doctoranda.\n\n"

    "Input: El senderista avanzaba observando el paisaje desde la ladera.\n"
    "Output: El senderista avanzaba observando el paisaje desde la ladera.\n\n"

    "Input: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n"
    "Output: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetLlama(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetLlama(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetLlama(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART y mT5
# para garantizar comparabilidad directa en la tabla del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Llama.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Llama:
    - Elimina comillas, guiones iniciales, "Corrección:" etc.
    - Llama Instruct a veces añade estas marcas incluso con
      la instrucción explícita de no hacerlo.
    """
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = re.sub(r'^(corrección|corregido|texto corregido)\s*:\s*',
                  '', text, flags=re.IGNORECASE)
    text = text.strip().strip('"\'')
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_llama(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Llama 3.2 3B")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_llama(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Llama 3.2 3B NIVEL 4 — Few-shot + reglas fuertes y conservadoras")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_llama(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Llama-3.2-3B-Instruct",
    "evaluation_mode":   "NIVEL 4 — Few-shot + reglas fuertes y conservadoras",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama.json'")
```

**FINE-TUNING**

```python
# ===============================================================
# PIPELINE FINE-TUNING GEC — Llama-3.2-3B-Instruct + QLoRA
# Unsloth 2026 — patrón oficial verificado
# ===============================================================
#
# INSTALACIÓN (ejecutar en celda separada antes de este script):
#   !pip install --upgrade "unsloth[colab-new]"
#
# STACK VERIFICADO:
#   unsloth >= 2026.3   — ya instalado en tu entorno
#   trl, transformers   — gestionados por Unsloth, no tocar
#
# ENMASCARAMIENTO DE PÉRDIDA:
#   train_on_responses_only de unsloth.chat_templates
#   Patrón oficial para Llama-3.x según docs.unsloth.ai
#   Calcula pérdida SOLO sobre tokens del assistant turn
#   Soluciona el overfit al prompt de versiones anteriores
# ===============================================================

# ---------------------------------------------------------------
# IMPORTANTE: unsloth SIEMPRE primero
# ---------------------------------------------------------------
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

import json, time, gc, re, random
import numpy as np
import torch
from torch.utils.data import Dataset as TorchDataset, DataLoader
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback
from datasets import Dataset as HFDataset

# ---------------------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------------------

SEED           = 42
MODEL_NAME     = "meta-llama/Llama-3.2-3B-Instruct"
MAX_SEQ_LENGTH = 2048
DTYPE          = None          # None = autodetección (bfloat16 en L4)
LOAD_IN_4BIT   = True

LORA_R         = 8
LORA_ALPHA     = 32
LORA_DROPOUT   = 0.0

NUM_EPOCHS     = 2
LEARNING_RATE  = 2e-5
BATCH_SIZE     = 2
GRAD_ACCUM     = 4             # batch efectivo = 8
WARMUP_STEPS   = 20
PATIENCE       = 1             # early stopping

MAX_NEW_TOKENS     = 128
REPETITION_PENALTY = 1.1
LLAMA3_EOS_IDS     = [128001, 128008, 128009]

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ---------------------------------------------------------------
# CARGA DEL MODELO
# ---------------------------------------------------------------

print("\n" + "=" * 60)
print("CARGANDO LLAMA-3.2-3B-INSTRUCT EN 4-BIT NF4 (QLoRA)")
print("=" * 60)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = DTYPE,
    load_in_4bit   = LOAD_IN_4BIT,
)

model = FastLanguageModel.get_peft_model(
    model,
    r              = LORA_R,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha     = LORA_ALPHA,
    lora_dropout   = LORA_DROPOUT,
    bias           = "none",
    use_gradient_checkpointing = "unsloth",
    random_state   = SEED,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

n_params       = sum(p.numel() for p in model.parameters())
n_params_train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parametros totales:      {n_params:,}")
print(f"Parametros entrenables:  {n_params_train:,} ({100*n_params_train/n_params:.2f}%)")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PROMPT
# ---------------------------------------------------------------

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal y en un único fenómeno: "
    "el gerundio de posterioridad.\n\n"

    "TAREA:\n"
    "Corrige SOLO los casos en los que el gerundio expresa una acción posterior a la del verbo principal.\n\n"

    "DEFINICIÓN:\n"
    "Hay gerundio de posterioridad únicamente cuando la acción expresada por el gerundio ocurre después de la acción del verbo principal. "
    "Si el gerundio expresa simultaneidad, anterioridad, modo, medio, causa, condición, descripción o forma parte de una perífrasis, no debe corregirse.\n\n"

    "OBJETIVO:\n"
    "Transforma esos casos en una coordinación con 'y' y un verbo conjugado.\n\n"

    "REGLA PRINCIPAL:\n"
    "Si hay duda sobre si el gerundio expresa posterioridad real, NO corrijas.\n\n"

    "CORREGIR SOLO SI:\n"
    "- el gerundio expresa con claridad una acción posterior;\n"
    "- la acción del gerundio ocurre después de la acción del verbo principal;\n"
    "- la corrección puede hacerse sustituyendo el gerundio por 'y' + verbo conjugado;\n"
    "- el nuevo verbo puede mantenerse en el mismo tiempo y modo que el verbo principal sin reformular libremente la oración.\n\n"

    "CÓMO CORREGIR:\n"
    "- Sustituye solo el segmento en gerundio afectado.\n"
    "- Usa 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo sujeto.\n"
    "- Ajusta pronombres o concordancia solo si es imprescindible para que la oración sea gramatical.\n"
    "- Conserva el significado original.\n\n"

    "NO CORREGIR NUNCA SI:\n"
    "- el gerundio expresa simultaneidad con el verbo principal;\n"
    "- el gerundio expresa anterioridad;\n"
    "- el gerundio expresa modo o manera: 'habló sonriendo', 'respondió mirando al suelo';\n"
    "- el gerundio expresa medio o instrumento;\n"
    "- el gerundio expresa causa, condición o descripción;\n"
    "- el gerundio forma parte de una perífrasis o construcción aspectual: 'está estudiando', 'sigue trabajando', 'continúa avanzando', 'acabó aceptando', 'terminó cediendo', 'lleva años investigando';\n"
    "- el gerundio aparece dentro de comillas, cita directa o estilo directo;\n"
    "- la frase no contiene gerundio;\n"
    "- la frase ya es correcta y no contiene gerundio de posterioridad;\n"
    "- la modificación exigiría reescribir libremente la oración o alterar partes ajenas al fenómeno.\n\n"

    "RESTRICCIONES DE SALIDA:\n"
    "- Devuelve una sola oración final.\n"
    "- No expliques nada.\n"
    "- No añadas comentarios.\n"
    "- No cambies palabras no relacionadas con el fenómeno.\n"
    "- No cambies léxico, nombres, números, puntuación, mayúsculas, comillas ni tildes, salvo lo imprescindible para la corrección mínima.\n"
    "- No resumas ni reformules.\n"
    "- Conserva exactamente el contenido original excepto los cambios mínimos necesarios para sustituir el gerundio de posterioridad.\n\n"

    "CRITERIO DE CONSERVACIÓN:\n"
    "Si solo una parte de la oración contiene un gerundio de posterioridad, corrige solo esa parte y deja intacto todo lo demás.\n\n"

    "IMPORTANTE:\n"
    "Es mejor dejar una frase sin corregir que hacer una corrección falsa."
)

USER_PROMPT = (
    "Ejemplos:\n\n"

    "Input: El laboratorio completó la primera fase del ensayo, publicando los resultados dos meses después.\n"
    "Output: El laboratorio completó la primera fase del ensayo y publicó los resultados dos meses después.\n\n"

    "Input: La comisaria inaugurará la exposición temporal, trasladándola después a la sede principal del museo.\n"
    "Output: La comisaria inaugurará la exposición temporal y la trasladará después a la sede principal del museo.\n\n"

    "Input: El decano presentó el nuevo plan docente, remitiéndolo al claustro al día siguiente para su valoración.\n"
    "Output: El decano presentó el nuevo plan docente y lo remitió al claustro al día siguiente para su valoración.\n\n"

    "Input: La empresa validó el prototipo final, registrando la patente semanas más tarde.\n"
    "Output: La empresa validó el prototipo final y registró la patente semanas más tarde.\n\n"

    "Input: El bibliotecario ordenará los manuscritos antiguos, digitalizándolos después para el archivo central.\n"
    "Output: El bibliotecario ordenará los manuscritos antiguos y los digitalizará después para el archivo central.\n\n"

    "Input: La arquitecta restauró la fachada principal, reabriendo el edificio meses después al público general.\n"
    "Output: La arquitecta restauró la fachada principal y reabrió el edificio meses después al público general.\n\n"

    "Input: El rector compareció ante la prensa, anunciando la decisión definitiva horas más tarde en un comunicado oficial.\n"
    "Output: El rector compareció ante la prensa y anunció la decisión definitiva horas más tarde en un comunicado oficial.\n\n"

    "Input: El equipo entregó la versión preliminar del informe, incorporando las correcciones la semana siguiente.\n"
    "Output: El equipo entregó la versión preliminar del informe e incorporó las correcciones la semana siguiente.\n\n"

    "Input: La profesora explicó el tema utilizando ejemplos del examen anterior.\n"
    "Output: La profesora explicó el tema utilizando ejemplos del examen anterior.\n\n"

    "Input: El violinista interpretó la pieza mirando al director en los compases finales.\n"
    "Output: El violinista interpretó la pieza mirando al director en los compases finales.\n\n"

    "Input: El departamento sigue revisando las solicitudes pendientes.\n"
    "Output: El departamento sigue revisando las solicitudes pendientes.\n\n"

    "Input: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n"
    "Output: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n\n"

    "Input: \"Estoy preparando la defensa\", respondió la doctoranda.\n"
    "Output: \"Estoy preparando la defensa\", respondió la doctoranda.\n\n"

    "Input: El senderista avanzaba observando el paisaje desde la ladera.\n"
    "Output: El senderista avanzaba observando el paisaje desde la ladera.\n\n"

    "Input: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n"
    "Output: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

EOS_TOKEN = tokenizer.eos_token

def build_prompt(input_text: str) -> list[dict]:
    return [
        {"role": "system",    "content": SYSTEM_PROMPT},
        {"role": "user",      "content": USER_PROMPT.format(sentence=input_text)},
    ]

def format_for_training(input_text: str, target_text: str) -> str:
    messages = [
        {"role": "system",    "content": SYSTEM_PROMPT},
        {"role": "user",      "content": USER_PROMPT.format(sentence=input_text)},
        {"role": "assistant", "content": target_text},
    ]
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    ) + EOS_TOKEN

# ---------------------------------------------------------------
# DIAGNÓSTICO DE TOKENS
# ---------------------------------------------------------------

def diagnostico_tokens(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        text = (
            format_for_training(row["input"], row["target"])
            if include_target
            else tokenizer.apply_chat_template(
                build_prompt(row["input"]),
                tokenize=False,
                add_generation_prompt=True,
            )
        )
        lengths.append(len(tokenizer(text, add_special_tokens=False)["input_ids"]))
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_SEQ_LENGTH).sum()
    print(f"\n--- Tokens: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_SEQ_LENGTH={MAX_SEQ_LENGTH}: "
          f"{truncated}/{len(lengths)}", end="")
    print(
        f"  ATENCION: sube MAX_SEQ_LENGTH a {int(lengths.max()) + 64}"
        if truncated > 0 else "  OK"
    )
    return lengths

print("\n" + "=" * 60)
print("DIAGNOSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens(df_train,          "TRAIN    (prompt + target)", include_target=True)
diagnostico_tokens(df_val,            "VAL      (prompt + target)", include_target=True)
diagnostico_tokens(df_test,           "TEST     (solo prompt)",     include_target=False)
diagnostico_tokens(df_noticias_final, "NOTICIAS (solo prompt)",     include_target=False)

# ---------------------------------------------------------------
# DATASET
# ---------------------------------------------------------------

def make_hf_dataset(df):
    return HFDataset.from_dict({
        "text": [
            format_for_training(row["input"], row["target"])
            for _, row in df.iterrows()
        ]
    })

train_dataset = make_hf_dataset(df_train)
val_dataset   = make_hf_dataset(df_val)

print(f"\nTrain: {len(train_dataset)} | Val: {len(val_dataset)}")
print(f"Batch efectivo: {BATCH_SIZE} x {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ---------------------------------------------------------------
# ENTRENAMIENTO
# Patrón oficial Unsloth 2026:
#   1. Crear SFTTrainer con SFTConfig
#   2. Envolver con train_on_responses_only (Llama-3.x tokens)
#   3. trainer.train()
# ---------------------------------------------------------------

trainer = SFTTrainer(
    model              = model,
    tokenizer          = tokenizer,
    train_dataset      = train_dataset,
    eval_dataset       = val_dataset,
    dataset_text_field = "text",
    max_seq_length     = MAX_SEQ_LENGTH,
    dataset_num_proc   = 2,
    packing            = False,
    args = SFTConfig(
        output_dir                  = "llama3b_gerundio_lora",
        per_device_train_batch_size = BATCH_SIZE,
        gradient_accumulation_steps = GRAD_ACCUM,
        num_train_epochs            = NUM_EPOCHS,
        learning_rate               = LEARNING_RATE,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.05,
        neftune_noise_alpha         = 5,
        fp16                        = not is_bfloat16_supported(),
        bf16                        = is_bfloat16_supported(),
        logging_steps               = 5,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        greater_is_better           = False,
        optim                       = "paged_adamw_8bit",
        seed                        = SEED,
        report_to                   = "none",
    ),
    callbacks = [EarlyStoppingCallback(early_stopping_patience=PATIENCE)],
)

# Enmascarar pérdida: solo tokens del assistant turn
# Patrón oficial Unsloth para Llama-3.1 / 3.2 / 3.3
trainer = train_on_responses_only(
    trainer,
    instruction_part = "<|start_header_id|>user<|end_header_id|>\n\n",
    response_part    = "<|start_header_id|>assistant<|end_header_id|>\n\n",
)

print("\nEmpieza entrenamiento Llama-3.2-3B-Instruct (QLoRA)")
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()
TRAIN_START = time.perf_counter()
trainer.train()
total_train_time = time.perf_counter() - TRAIN_START
print(f"Entrenamiento completado en {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
if device == "cuda":
    print(f"Pico de memoria: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PREPARAR PARA INFERENCIA
#
# NO usar FastLanguageModel.for_inference(model):
# reemplaza las tablas RoPE con valores hardcodeados a la
# longitud de entrenamiento, causando RuntimeError cuando
# los prompts de inferencia son más largos que esa longitud.
# Con prompts de ~1500 tokens y MAX_SEQ_LENGTH=2048, el offset
# de posición durante inferencia supera el límite compilado.
# model.eval() es correcto y seguro en este escenario.
# ---------------------------------------------------------------

model.eval()
tokenizer.padding_side = "left"

# ---------------------------------------------------------------
# MÉTRICAS GEC
# ---------------------------------------------------------------

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    text = text.strip()
    text = re.sub(
        r'^(correccion|corrección|corregido|output|respuesta|sentence)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-]+', '', text).strip()
    text = text.strip('"\'').strip()
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    # Si el modelo genera razonamiento en inglés, devolver vacío
    if re.match(
        r'^(if |when |the |this |copy |there |sentence |input |output )',
        text, flags=re.IGNORECASE
    ):
        return ""
    return text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    if not pred_text:
        pred_text = src_text

    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()
    gleu             = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta    = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )
    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}
    n  = len(all_metrics)
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05       = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )
    return {
        "n_total":           n,
        "n_positive":        tp + fn,
        "n_negative":        tn + fp,
        "tp": tp, "tn": tn, "fp": fp, "fn": fn,
        "accuracy":          accuracy,
        "precision":         precision,
        "recall":            recall,
        "f05":               f05,
        "gleu_mean":         float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":          float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":      float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":       float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":     float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":        float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":      float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":   float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS -- {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc, pm = p["needs_correction"], p["pred_modified"]
        if   nc and     pm: label = "TP  (con errores -> modelo corrige)"
        elif nc and not pm: label = "FN  (con errores -> modelo no corrige)"
        elif not nc and pm: label = "FP  (sin errores -> modelo modifica igualmente)"
        else:               label = "TN  (sin errores -> modelo no toca)"
        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} | ED_norm={p['ed_norm']:.3f}")


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"METRICAS AGREGADAS -- {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  -- Deteccion (clasificacion binaria) --")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  (frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  (de las que corrige, cuantas tenian error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  (de las que tenian error, cuantas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  (penaliza mas FP que FN, estandar GEC)")
    print()
    print(f"  -- Calidad de correccion (token-level) --")
    print(f"     GLEU:           {agg.get('gleu_mean',    0):.4f} +/- {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean', 0):.4f} +/- {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  -- Comportamiento del modelo --")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  -- Tiempos --")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ---------------------------------------------------------------
# EVALUACIÓN
# ---------------------------------------------------------------

class GECDataset(TorchDataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()
    def __len__(self):  return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

test_loader     = DataLoader(GECDataset(df_test),           batch_size=1, shuffle=False)
noticias_loader = DataLoader(GECDataset(df_noticias_final), batch_size=1, shuffle=False)
print(f"\nTest: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")


def evaluate_model(loader, src_texts_list=None):
    model.eval()
    all_metrics, predictions = [], []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            enc = tokenizer.apply_chat_template(
                build_prompt(input_text),
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids          = input_ids,
                attention_mask     = attention_mask,
                max_new_tokens     = MAX_NEW_TOKENS,
                do_sample          = False,
                repetition_penalty = REPETITION_PENALTY,
                use_cache          = False,
                pad_token_id       = tokenizer.eos_token_id,
                eos_token_id       = LLAMA3_EOS_IDS,
            )

            pred_raw  = decode_pred(outputs[0][input_length:])
            pred_text = clean_pred(pred_raw)
            src_text  = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src": src_text, "tgt": tgt_text,
                "pred_raw": pred_raw, "pred": pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions


# TEST
print("\n" + "=" * 60)
print("EVALUACION EN TEST SET")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

t0 = time.perf_counter()
test_agg, test_preds = evaluate_model(test_loader, df_test["input"].tolist())
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - t0

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time, len(test_preds), "TEST SET")

# NOTICIAS
print("\n" + "=" * 60)
print("EVALUACION EN NOTICIAS")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

t0 = time.perf_counter()
noticias_agg, noticias_preds = evaluate_model(
    noticias_loader, df_noticias_final["input"].tolist()
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - t0

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time, len(noticias_preds), "NOTICIAS")

# ---------------------------------------------------------------
# RESULTADOS FINALES
# ---------------------------------------------------------------

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":          "Llama-3.2-3B-Instruct",
    "evaluation_mode":     "fine-tuned QLoRA",
    "seed":                SEED,
    "device":              device,
    "n_params":            n_params,
    "n_params_trainable":  n_params_train,
    "quantization":        "nf4_4bit",
    "lora_r":              LORA_R,
    "lora_alpha":          LORA_ALPHA,
    "lora_dropout":        LORA_DROPOUT,
    "epochs":              NUM_EPOCHS,
    "learning_rate":       LEARNING_RATE,
    "warmup_steps":        WARMUP_STEPS,
    "patience":            PATIENCE,
    "neftune_noise_alpha": 5,
    "batch_size":          BATCH_SIZE,
    "grad_accum":          GRAD_ACCUM,
    "effective_batch":     BATCH_SIZE * GRAD_ACCUM,
    "max_seq_length":      MAX_SEQ_LENGTH,
    "max_new_tokens":      MAX_NEW_TOKENS,
    "do_sample":           False,
    "repetition_penalty":  REPETITION_PENALTY,
    "decode_strategy":     "greedy",
    "loss_masking":        "train_on_responses_only",
    "total_train_time_s":  total_train_time,

    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES -- RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama3b_finetuned_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama3b_finetuned_gerundio.json'")
```

```python
# ===============================================================
# PIPELINE FINE-TUNING GEC — Llama-3.2-3B-Instruct + QLoRA
# Unsloth 2026 — patrón oficial verificado
# ===============================================================
#
# INSTALACIÓN (ejecutar en celda separada antes de este script):
#   !pip install --upgrade "unsloth[colab-new]"
#
# STACK VERIFICADO:
#   unsloth >= 2026.3   — ya instalado en tu entorno
#   trl, transformers   — gestionados por Unsloth, no tocar
#
# ENMASCARAMIENTO DE PÉRDIDA:
#   train_on_responses_only de unsloth.chat_templates
#   Patrón oficial para Llama-3.x según docs.unsloth.ai
#   Calcula pérdida SOLO sobre tokens del assistant turn
#   Soluciona el overfit al prompt de versiones anteriores
# ===============================================================

# ---------------------------------------------------------------
# IMPORTANTE: unsloth SIEMPRE primero
# ---------------------------------------------------------------
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

import json, time, gc, re, random
import numpy as np
import torch
from torch.utils.data import Dataset as TorchDataset, DataLoader
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback
from datasets import Dataset as HFDataset

# ---------------------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------------------

SEED           = 42
MODEL_NAME     = "meta-llama/Llama-3.2-3B-Instruct"
MAX_SEQ_LENGTH = 2048
DTYPE          = None          # None = autodetección (bfloat16 en L4)
LOAD_IN_4BIT   = True

LORA_R         = 8
LORA_ALPHA     = 32
LORA_DROPOUT   = 0.0

NUM_EPOCHS     = 2
LEARNING_RATE  = 2e-5
BATCH_SIZE     = 2
GRAD_ACCUM     = 4             # batch efectivo = 8
WARMUP_STEPS   = 20
PATIENCE       = 1             # early stopping

MAX_NEW_TOKENS     = 128
REPETITION_PENALTY = 1.1
LLAMA3_EOS_IDS     = [128001, 128008, 128009]

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ---------------------------------------------------------------
# CARGA DEL MODELO
# ---------------------------------------------------------------

print("\n" + "=" * 60)
print("CARGANDO LLAMA-3.2-3B-INSTRUCT EN 4-BIT NF4 (QLoRA)")
print("=" * 60)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = DTYPE,
    load_in_4bit   = LOAD_IN_4BIT,
)

model = FastLanguageModel.get_peft_model(
    model,
    r              = LORA_R,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha     = LORA_ALPHA,
    lora_dropout   = LORA_DROPOUT,
    bias           = "none",
    use_gradient_checkpointing = "unsloth",
    random_state   = SEED,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

n_params       = sum(p.numel() for p in model.parameters())
n_params_train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parametros totales:      {n_params:,}")
print(f"Parametros entrenables:  {n_params_train:,} ({100*n_params_train/n_params:.2f}%)")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PROMPT
# ---------------------------------------------------------------

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Devuelve solo la oración final."
)

USER_PROMPT = (
    "Input: La investigadora cerró la fase de recogida de datos, redactando el informe semanas después.\n"
    "Output: La investigadora cerró la fase de recogida de datos y redactó el informe semanas después.\n\n"

    "Input: El técnico revisará el servidor principal, reiniciándolo más tarde para aplicar la configuración definitiva.\n"
    "Output: El técnico revisará el servidor principal y lo reiniciará más tarde para aplicar la configuración definitiva.\n\n"

    "Input: El museo restauró la pieza dañada, exhibiéndola meses después en una sala temporal.\n"
    "Output: El museo restauró la pieza dañada y la exhibió meses después en una sala temporal.\n\n"

    "Input: La abogada preparó la documentación, presentándola al día siguiente en el registro.\n"
    "Output: La abogada preparó la documentación y la presentó al día siguiente en el registro.\n\n"

    "Input: El alumnado siguió trabajando en silencio durante la prueba.\n"
    "Output: El alumnado siguió trabajando en silencio durante la prueba.\n\n"

    "Input: El conferenciante habló sonriendo al público durante toda la sesión.\n"
    "Output: El conferenciante habló sonriendo al público durante toda la sesión.\n\n"

    "Input: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n"
    "Output: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n\n"

    "Input: El equipo terminó aceptando la propuesta tras varias horas de debate.\n"
    "Output: El equipo terminó aceptando la propuesta tras varias horas de debate.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

EOS_TOKEN = tokenizer.eos_token

def build_prompt(input_text: str) -> list[dict]:
    return [
        {"role": "system",    "content": SYSTEM_PROMPT},
        {"role": "user",      "content": USER_PROMPT.format(sentence=input_text)},
    ]

def format_for_training(input_text: str, target_text: str) -> str:
    messages = [
        {"role": "system",    "content": SYSTEM_PROMPT},
        {"role": "user",      "content": USER_PROMPT.format(sentence=input_text)},
        {"role": "assistant", "content": target_text},
    ]
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    ) + EOS_TOKEN

# ---------------------------------------------------------------
# DIAGNÓSTICO DE TOKENS
# ---------------------------------------------------------------

def diagnostico_tokens(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        text = (
            format_for_training(row["input"], row["target"])
            if include_target
            else tokenizer.apply_chat_template(
                build_prompt(row["input"]),
                tokenize=False,
                add_generation_prompt=True,
            )
        )
        lengths.append(len(tokenizer(text, add_special_tokens=False)["input_ids"]))
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_SEQ_LENGTH).sum()
    print(f"\n--- Tokens: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_SEQ_LENGTH={MAX_SEQ_LENGTH}: "
          f"{truncated}/{len(lengths)}", end="")
    print(
        f"  ATENCION: sube MAX_SEQ_LENGTH a {int(lengths.max()) + 64}"
        if truncated > 0 else "  OK"
    )
    return lengths

print("\n" + "=" * 60)
print("DIAGNOSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens(df_train,          "TRAIN    (prompt + target)", include_target=True)
diagnostico_tokens(df_val,            "VAL      (prompt + target)", include_target=True)
diagnostico_tokens(df_test,           "TEST     (solo prompt)",     include_target=False)
diagnostico_tokens(df_noticias_final, "NOTICIAS (solo prompt)",     include_target=False)

# ---------------------------------------------------------------
# DATASET
# ---------------------------------------------------------------

def make_hf_dataset(df):
    return HFDataset.from_dict({
        "text": [
            format_for_training(row["input"], row["target"])
            for _, row in df.iterrows()
        ]
    })

train_dataset = make_hf_dataset(df_train)
val_dataset   = make_hf_dataset(df_val)

print(f"\nTrain: {len(train_dataset)} | Val: {len(val_dataset)}")
print(f"Batch efectivo: {BATCH_SIZE} x {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ---------------------------------------------------------------
# ENTRENAMIENTO
# Patrón oficial Unsloth 2026:
#   1. Crear SFTTrainer con SFTConfig
#   2. Envolver con train_on_responses_only (Llama-3.x tokens)
#   3. trainer.train()
# ---------------------------------------------------------------

trainer = SFTTrainer(
    model              = model,
    tokenizer          = tokenizer,
    train_dataset      = train_dataset,
    eval_dataset       = val_dataset,
    dataset_text_field = "text",
    max_seq_length     = MAX_SEQ_LENGTH,
    dataset_num_proc   = 2,
    packing            = False,
    args = SFTConfig(
        output_dir                  = "llama3b_gerundio_lora",
        per_device_train_batch_size = BATCH_SIZE,
        gradient_accumulation_steps = GRAD_ACCUM,
        num_train_epochs            = NUM_EPOCHS,
        learning_rate               = LEARNING_RATE,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.05,
        neftune_noise_alpha         = 5,
        fp16                        = not is_bfloat16_supported(),
        bf16                        = is_bfloat16_supported(),
        logging_steps               = 5,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        greater_is_better           = False,
        optim                       = "paged_adamw_8bit",
        seed                        = SEED,
        report_to                   = "none",
    ),
    callbacks = [EarlyStoppingCallback(early_stopping_patience=PATIENCE)],
)

# Enmascarar pérdida: solo tokens del assistant turn
# Patrón oficial Unsloth para Llama-3.1 / 3.2 / 3.3
trainer = train_on_responses_only(
    trainer,
    instruction_part = "<|start_header_id|>user<|end_header_id|>\n\n",
    response_part    = "<|start_header_id|>assistant<|end_header_id|>\n\n",
)

print("\nEmpieza entrenamiento Llama-3.2-3B-Instruct (QLoRA)")
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()
TRAIN_START = time.perf_counter()
trainer.train()
total_train_time = time.perf_counter() - TRAIN_START
print(f"Entrenamiento completado en {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
if device == "cuda":
    print(f"Pico de memoria: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PREPARAR PARA INFERENCIA
#
# NO usar FastLanguageModel.for_inference(model):
# reemplaza las tablas RoPE con valores hardcodeados a la
# longitud de entrenamiento, causando RuntimeError cuando
# los prompts de inferencia son más largos que esa longitud.
# Con prompts de ~1500 tokens y MAX_SEQ_LENGTH=2048, el offset
# de posición durante inferencia supera el límite compilado.
# model.eval() es correcto y seguro en este escenario.
# ---------------------------------------------------------------

model.eval()
tokenizer.padding_side = "left"

# ---------------------------------------------------------------
# MÉTRICAS GEC
# ---------------------------------------------------------------

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    text = text.strip()
    text = re.sub(
        r'^(correccion|corrección|corregido|output|respuesta|sentence)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-]+', '', text).strip()
    text = text.strip('"\'').strip()
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    # Si el modelo genera razonamiento en inglés, devolver vacío
    if re.match(
        r'^(if |when |the |this |copy |there |sentence |input |output )',
        text, flags=re.IGNORECASE
    ):
        return ""
    return text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    if not pred_text:
        pred_text = src_text

    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()
    gleu             = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta    = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )
    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}
    n  = len(all_metrics)
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05       = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )
    return {
        "n_total":           n,
        "n_positive":        tp + fn,
        "n_negative":        tn + fp,
        "tp": tp, "tn": tn, "fp": fp, "fn": fn,
        "accuracy":          accuracy,
        "precision":         precision,
        "recall":            recall,
        "f05":               f05,
        "gleu_mean":         float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":          float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":      float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":       float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":     float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":        float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":      float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":   float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS -- {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc, pm = p["needs_correction"], p["pred_modified"]
        if   nc and     pm: label = "TP  (con errores -> modelo corrige)"
        elif nc and not pm: label = "FN  (con errores -> modelo no corrige)"
        elif not nc and pm: label = "FP  (sin errores -> modelo modifica igualmente)"
        else:               label = "TN  (sin errores -> modelo no toca)"
        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} | ED_norm={p['ed_norm']:.3f}")


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"METRICAS AGREGADAS -- {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  -- Deteccion (clasificacion binaria) --")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  (frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  (de las que corrige, cuantas tenian error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  (de las que tenian error, cuantas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  (penaliza mas FP que FN, estandar GEC)")
    print()
    print(f"  -- Calidad de correccion (token-level) --")
    print(f"     GLEU:           {agg.get('gleu_mean',    0):.4f} +/- {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean', 0):.4f} +/- {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  -- Comportamiento del modelo --")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  -- Tiempos --")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ---------------------------------------------------------------
# EVALUACIÓN
# ---------------------------------------------------------------

class GECDataset(TorchDataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()
    def __len__(self):  return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

test_loader     = DataLoader(GECDataset(df_test),           batch_size=1, shuffle=False)
noticias_loader = DataLoader(GECDataset(df_noticias_final), batch_size=1, shuffle=False)
print(f"\nTest: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")


def evaluate_model(loader, src_texts_list=None):
    model.eval()
    all_metrics, predictions = [], []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            enc = tokenizer.apply_chat_template(
                build_prompt(input_text),
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids          = input_ids,
                attention_mask     = attention_mask,
                max_new_tokens     = MAX_NEW_TOKENS,
                do_sample          = False,
                repetition_penalty = REPETITION_PENALTY,
                use_cache          = False,
                pad_token_id       = tokenizer.eos_token_id,
                eos_token_id       = LLAMA3_EOS_IDS,
            )

            pred_raw  = decode_pred(outputs[0][input_length:])
            pred_text = clean_pred(pred_raw)
            src_text  = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src": src_text, "tgt": tgt_text,
                "pred_raw": pred_raw, "pred": pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions


# TEST
print("\n" + "=" * 60)
print("EVALUACION EN TEST SET")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

t0 = time.perf_counter()
test_agg, test_preds = evaluate_model(test_loader, df_test["input"].tolist())
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - t0

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time, len(test_preds), "TEST SET")

# NOTICIAS
print("\n" + "=" * 60)
print("EVALUACION EN NOTICIAS")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

t0 = time.perf_counter()
noticias_agg, noticias_preds = evaluate_model(
    noticias_loader, df_noticias_final["input"].tolist()
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - t0

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time, len(noticias_preds), "NOTICIAS")

# ---------------------------------------------------------------
# RESULTADOS FINALES
# ---------------------------------------------------------------

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":          "Llama-3.2-3B-Instruct",
    "evaluation_mode":     "fine-tuned QLoRA",
    "seed":                SEED,
    "device":              device,
    "n_params":            n_params,
    "n_params_trainable":  n_params_train,
    "quantization":        "nf4_4bit",
    "lora_r":              LORA_R,
    "lora_alpha":          LORA_ALPHA,
    "lora_dropout":        LORA_DROPOUT,
    "epochs":              NUM_EPOCHS,
    "learning_rate":       LEARNING_RATE,
    "warmup_steps":        WARMUP_STEPS,
    "patience":            PATIENCE,
    "neftune_noise_alpha": 5,
    "batch_size":          BATCH_SIZE,
    "grad_accum":          GRAD_ACCUM,
    "effective_batch":     BATCH_SIZE * GRAD_ACCUM,
    "max_seq_length":      MAX_SEQ_LENGTH,
    "max_new_tokens":      MAX_NEW_TOKENS,
    "do_sample":           False,
    "repetition_penalty":  REPETITION_PENALTY,
    "decode_strategy":     "greedy",
    "loss_masking":        "train_on_responses_only",
    "total_train_time_s":  total_train_time,

    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES -- RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_llama3b_finetuned_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_llama3b_finetuned_gerundio.json'")
```

# **MISTRAL**

**PROMPT 1**

```python
# ============================================================
# PIPELINE GEC — Mistral-7B-Instruct-v0.3 (zero-shot)
# Solo inferencia: Mistral se evalúa sin fine-tuning.
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 7B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# max_new_tokens=150: idéntico a Llama para comparabilidad.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~55 tokens
#   de Mistral (tokenizador SentencePiece). 150 deja margen
#   suficiente sin truncar en ambos modelos.
#
# repetition_penalty=1.00: idéntico a Llama.
#   Mismo valor para garantizar que esta variable no introduce
#   diferencias entre modelos en la comparación. Cualquier
#   diferencia en los resultados es atribuible al modelo,
#   no a la configuración de generación.
#
# padding_side="left": necesario para Mistral específicamente.
#   El tokenizador de Mistral-7B no configura padding_side
#   por defecto, a diferencia del de Llama. Se fija "left"
#   porque es el comportamiento correcto para modelos causales.
#   Con batch_size=1 no afecta a los resultados, pero garantiza
#   que el pipeline es reutilizable sin modificación.
#
# torch_dtype=bfloat16: necesario para Mistral específicamente.
#   Mistral 7B sin este parámetro inicializa las capas no
#   cuantizadas (embedding, lm_head) en fp32. Se especifica
#   explícitamente para alinearlo con compute_dtype=bfloat16
#   del BitsAndBytesConfig. En Llama 3B device_map="auto"
#   ya gestiona esto correctamente sin necesidad de fijarlo.
#
# COMPARABILIDAD:
#   Mistral se evalúa zero-shot (sin fine-tuning), igual que
#   Llama. Los parámetros de generación (do_sample, num_beams,
#   max_new_tokens, repetition_penalty) son idénticos a Llama
#   para garantizar comparabilidad directa en la tabla del paper.
#   Las diferencias estructurales (padding_side, torch_dtype)
#   son requisitos de la arquitectura de Mistral y no alteran
#   el resultado de las métricas ni la validez de la comparación.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Mistral 7B de ~14GB (bf16) a ~4.2GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.5GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# torch_dtype=bfloat16: alinea las capas no cuantizadas
#   (embedding, lm_head) con compute_dtype. Ver sección 0.
#
# device_map="auto": distribuye capas automáticamente en GPU.
#
# padding_side="left": requerido por la arquitectura de
#   Mistral para generación causal correcta. Ver sección 0.
# ============================================================

print("="*60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Cuando el gerundio exprese una acción posterior a la del verbo principal, sustitúyelo por una construcción con 'y' y verbo conjugado. "
    "No modifiques nada más. "
    "Devuelve solo la oración final. "
    "Si no es necesario corregir, devuelve la oración exactamente igual."
)

USER_PROMPT = "{sentence}"


def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetMistral(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetMistral(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetMistral(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART, mT5 y
# Llama para garantizar comparabilidad directa en la tabla
# del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Mistral.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Mistral.
    Mistral Instruct puede añadir prefijos o comillas incluso
    con la instrucción explícita de no hacerlo. Se amplía
    respecto a Llama con 'output' y 'respuesta' porque el
    chat template de Mistral puede introducirlos. La lógica
    de extracción de primera oración es idéntica a Llama.
    """
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_mistral(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Mistral-7B-Instruct-v0.3 (zero-shot)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_mistral(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Mistral-7B-Instruct-v0.3 (zero-shot)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_mistral(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":   "zero_shot",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral_zeroshot_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral_zeroshot_gerundio.json'")
```

**PROMPT 2**

```python
# ============================================================
# PIPELINE GEC — Mistral-7B-Instruct-v0.3
# Solo inferencia: Mistral se evalúa sin fine-tuning.
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 7B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# max_new_tokens=150: idéntico a Llama para comparabilidad.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~55 tokens
#   de Mistral (tokenizador SentencePiece). 150 deja margen
#   suficiente sin truncar en ambos modelos.
#
# repetition_penalty=1.00: idéntico a Llama.
#   Mismo valor para garantizar que esta variable no introduce
#   diferencias entre modelos en la comparación. Cualquier
#   diferencia en los resultados es atribuible al modelo,
#   no a la configuración de generación.
#
# padding_side="left": necesario para Mistral específicamente.
#   El tokenizador de Mistral-7B no configura padding_side
#   por defecto, a diferencia del de Llama. Se fija "left"
#   porque es el comportamiento correcto para modelos causales.
#   Con batch_size=1 no afecta a los resultados, pero garantiza
#   que el pipeline es reutilizable sin modificación.
#
# torch_dtype=bfloat16: necesario para Mistral específicamente.
#   Mistral 7B sin este parámetro inicializa las capas no
#   cuantizadas (embedding, lm_head) en fp32. Se especifica
#   explícitamente para alinearlo con compute_dtype=bfloat16
#   del BitsAndBytesConfig. En Llama 3B device_map="auto"
#   ya gestiona esto correctamente sin necesidad de fijarlo.
#
# COMPARABILIDAD:
#   Mistral se evalúa zero-shot (sin fine-tuning), igual que
#   Llama. Los parámetros de generación (do_sample, num_beams,
#   max_new_tokens, repetition_penalty) son idénticos a Llama
#   para garantizar comparabilidad directa en la tabla del paper.
#   Las diferencias estructurales (padding_side, torch_dtype)
#   son requisitos de la arquitectura de Mistral y no alteran
#   el resultado de las métricas ni la validez de la comparación.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Mistral 7B de ~14GB (bf16) a ~4.2GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.5GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# torch_dtype=bfloat16: alinea las capas no cuantizadas
#   (embedding, lm_head) con compute_dtype. Ver sección 0.
#
# device_map="auto": distribuye capas automáticamente en GPU.
#
# padding_side="left": requerido por la arquitectura de
#   Mistral para generación causal correcta. Ver sección 0.
# ============================================================

print("="*60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal.\n\n"

    "TAREA:\n"
    "Detecta y corrige únicamente el gerundio de posterioridad.\n\n"

    "REGLA PRINCIPAL:\n"
    "Existe gerundio de posterioridad cuando el gerundio expresa una acción que ocurre después de la acción del verbo principal. "
    "En esos casos, debe reformularse mediante una coordinación con 'y' y un verbo conjugado.\n\n"

    "CÓMO CORREGIR:\n"
    "- Corrige solo el gerundio de posterioridad.\n"
    "- Sustituye el gerundio por 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo significado.\n"
    "- Ajusta pronombres o concordancia solo si es estrictamente necesario.\n"
    "- Modifica solo lo mínimo imprescindible.\n"
    "- No reformules la oración.\n\n"

    "NO CORREGIR EN ESTOS CASOS:\n"
    "- Si el gerundio expresa simultaneidad.\n"
    "- Si el gerundio expresa modo, medio, causa, condición o descripción.\n"
    "- Si forma parte de una perífrasis verbal o construcción aspectual (por ejemplo: 'está estudiando', 'sigue trabajando', 'terminó aceptando', 'acabó cediendo').\n"
    "- Si aparece dentro de comillas o estilo directo.\n"
    "- Si no hay gerundio.\n"
    "- Si el fenómeno no está presente.\n\n"

    "IMPORTANTE:\n"
    "- No cambies palabras innecesariamente.\n"
    "- No mejores el estilo.\n"
    "- No añadas ni elimines información.\n"
    "- Si no estás seguro de que el gerundio expresa posterioridad, no hagas cambios.\n\n"

    "SALIDA:\n"
    "Devuelve únicamente la oración final corregida. "
    "Si no es necesario corregir, devuelve la oración exactamente igual."
)

USER_PROMPT = "{sentence}"


def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetMistral(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetMistral(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetMistral(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART, mT5 y
# Llama para garantizar comparabilidad directa en la tabla
# del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Mistral.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Mistral.
    Mistral Instruct puede añadir prefijos o comillas incluso
    con la instrucción explícita de no hacerlo. Se amplía
    respecto a Llama con 'output' y 'respuesta' porque el
    chat template de Mistral puede introducirlos. La lógica
    de extracción de primera oración es idéntica a Llama.
    """
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_mistral(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Mistral-7B-Instruct-v0.3 (NIVEL 2 — Zero-shot guiado con reglas claras)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_mistral(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Mistral-7B-Instruct-v0.3 (zero-shot)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_mistral(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":   "NIVEL 2 — Zero-shot guiado con reglas claras",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral_zeroshot_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral_zeroshot_gerundio.json'")
```

**PROMPT 3**

```python
# ============================================================
# PIPELINE GEC — Mistral-7B-Instruct-v0.3
# Solo inferencia: Mistral se evalúa sin fine-tuning.
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 7B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# max_new_tokens=150: idéntico a Llama para comparabilidad.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~55 tokens
#   de Mistral (tokenizador SentencePiece). 150 deja margen
#   suficiente sin truncar en ambos modelos.
#
# repetition_penalty=1.00: idéntico a Llama.
#   Mismo valor para garantizar que esta variable no introduce
#   diferencias entre modelos en la comparación. Cualquier
#   diferencia en los resultados es atribuible al modelo,
#   no a la configuración de generación.
#
# padding_side="left": necesario para Mistral específicamente.
#   El tokenizador de Mistral-7B no configura padding_side
#   por defecto, a diferencia del de Llama. Se fija "left"
#   porque es el comportamiento correcto para modelos causales.
#   Con batch_size=1 no afecta a los resultados, pero garantiza
#   que el pipeline es reutilizable sin modificación.
#
# torch_dtype=bfloat16: necesario para Mistral específicamente.
#   Mistral 7B sin este parámetro inicializa las capas no
#   cuantizadas (embedding, lm_head) en fp32. Se especifica
#   explícitamente para alinearlo con compute_dtype=bfloat16
#   del BitsAndBytesConfig. En Llama 3B device_map="auto"
#   ya gestiona esto correctamente sin necesidad de fijarlo.
#
# COMPARABILIDAD:
#   Mistral se evalúa zero-shot (sin fine-tuning), igual que
#   Llama. Los parámetros de generación (do_sample, num_beams,
#   max_new_tokens, repetition_penalty) son idénticos a Llama
#   para garantizar comparabilidad directa en la tabla del paper.
#   Las diferencias estructurales (padding_side, torch_dtype)
#   son requisitos de la arquitectura de Mistral y no alteran
#   el resultado de las métricas ni la validez de la comparación.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Mistral 7B de ~14GB (bf16) a ~4.2GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.5GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# torch_dtype=bfloat16: alinea las capas no cuantizadas
#   (embedding, lm_head) con compute_dtype. Ver sección 0.
#
# device_map="auto": distribuye capas automáticamente en GPU.
#
# padding_side="left": requerido por la arquitectura de
#   Mistral para generación causal correcta. Ver sección 0.
# ============================================================

print("="*60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Devuelve solo la oración final."
)

USER_PROMPT = (
    "Input: La investigadora cerró la fase de recogida de datos, redactando el informe semanas después.\n"
    "Output: La investigadora cerró la fase de recogida de datos y redactó el informe semanas después.\n\n"

    "Input: El técnico revisará el servidor principal, reiniciándolo más tarde para aplicar la configuración definitiva.\n"
    "Output: El técnico revisará el servidor principal y lo reiniciará más tarde para aplicar la configuración definitiva.\n\n"

    "Input: El museo restauró la pieza dañada, exhibiéndola meses después en una sala temporal.\n"
    "Output: El museo restauró la pieza dañada y la exhibió meses después en una sala temporal.\n\n"

    "Input: La abogada preparó la documentación, presentándola al día siguiente en el registro.\n"
    "Output: La abogada preparó la documentación y la presentó al día siguiente en el registro.\n\n"

    "Input: El alumnado siguió trabajando en silencio durante la prueba.\n"
    "Output: El alumnado siguió trabajando en silencio durante la prueba.\n\n"

    "Input: El conferenciante habló sonriendo al público durante toda la sesión.\n"
    "Output: El conferenciante habló sonriendo al público durante toda la sesión.\n\n"

    "Input: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n"
    "Output: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n\n"

    "Input: El equipo terminó aceptando la propuesta tras varias horas de debate.\n"
    "Output: El equipo terminó aceptando la propuesta tras varias horas de debate.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetMistral(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetMistral(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetMistral(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART, mT5 y
# Llama para garantizar comparabilidad directa en la tabla
# del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Mistral.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Mistral.
    Mistral Instruct puede añadir prefijos o comillas incluso
    con la instrucción explícita de no hacerlo. Se amplía
    respecto a Llama con 'output' y 'respuesta' porque el
    chat template de Mistral puede introducirlos. La lógica
    de extracción de primera oración es idéntica a Llama.
    """
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_mistral(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Mistral-7B-Instruct-v0.3 (NIVEL 3: Few-shot con pocas regla)")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_mistral(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Mistral-7B-Instruct-v0.3")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_mistral(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":   "NIVEL 3 — Few-shot con pocas regla",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral_zeroshot_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral_gerundio.json'")
```

**PROMPT 4**

```python
# ============================================================
# PIPELINE GEC — Mistral-7B-Instruct-v0.3
# Solo inferencia: Mistral se evalúa sin fine-tuning.
# ============================================================

import json, time, gc, random, re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter

# ============================================================
# 0. CONFIGURACIÓN CENTRALIZADA
#
# do_sample=False: inferencia determinista (greedy decoding).
#   Con do_sample=False temperature no tiene efecto y se ignora.
#   Para GEC queremos la salida más probable, no muestreo
#   estocástico. Resultados reproducibles entre ejecuciones.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# num_beams=1: greedy decoding en lugar de beam search.
#   Beam search en LLMs causales de 7B multiplica el KV-cache
#   por num_beams. Con 4-bit NF4 en L4 (23.6GB), num_beams>1
#   aumenta tiempo y memoria sin beneficio claro en zero-shot.
#   Greedy es el estándar en evaluación zero-shot de LLMs.
#   Idéntico a Llama: garantiza comparabilidad directa.
#
# max_new_tokens=150: idéntico a Llama para comparabilidad.
#   El dataset tiene frases de hasta ~40 palabras ≈ ~55 tokens
#   de Mistral (tokenizador SentencePiece). 150 deja margen
#   suficiente sin truncar en ambos modelos.
#
# repetition_penalty=1.00: idéntico a Llama.
#   Mismo valor para garantizar que esta variable no introduce
#   diferencias entre modelos en la comparación. Cualquier
#   diferencia en los resultados es atribuible al modelo,
#   no a la configuración de generación.
#
# padding_side="left": necesario para Mistral específicamente.
#   El tokenizador de Mistral-7B no configura padding_side
#   por defecto, a diferencia del de Llama. Se fija "left"
#   porque es el comportamiento correcto para modelos causales.
#   Con batch_size=1 no afecta a los resultados, pero garantiza
#   que el pipeline es reutilizable sin modificación.
#
# torch_dtype=bfloat16: necesario para Mistral específicamente.
#   Mistral 7B sin este parámetro inicializa las capas no
#   cuantizadas (embedding, lm_head) en fp32. Se especifica
#   explícitamente para alinearlo con compute_dtype=bfloat16
#   del BitsAndBytesConfig. En Llama 3B device_map="auto"
#   ya gestiona esto correctamente sin necesidad de fijarlo.
#
# COMPARABILIDAD:
#   Mistral se evalúa zero-shot (sin fine-tuning), igual que
#   Llama. Los parámetros de generación (do_sample, num_beams,
#   max_new_tokens, repetition_penalty) son idénticos a Llama
#   para garantizar comparabilidad directa en la tabla del paper.
#   Las diferencias estructurales (padding_side, torch_dtype)
#   son requisitos de la arquitectura de Mistral y no alteran
#   el resultado de las métricas ni la validez de la comparación.
#   La diferencia zero-shot vs fine-tuned se documenta en el JSON.
# ============================================================

SEED               = 42
MODEL_NAME         = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_NEW_TOKENS     = 150
DO_SAMPLE          = False
NUM_BEAMS          = 1
REPETITION_PENALTY = 1.00

# ============================================================
# 1. REPRODUCIBILIDAD
# ============================================================

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

def clear_cache():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ============================================================
# 2. CARGA DEL MODELO EN 4-BIT NF4
#
# NF4 (Normal Float 4): cuantización de 4 bits optimizada para
#   distribuciones de pesos de modelos preentrenados.
#   Reduce Mistral 7B de ~14GB (bf16) a ~4.2GB.
#
# double_quant=True: cuantiza también las constantes de
#   cuantización, ahorrando ~0.5GB adicionales.
#
# compute_dtype=bfloat16: las operaciones de atención y FFN
#   se ejecutan en bf16 aunque los pesos estén en nf4.
#   bf16 > fp16 para estabilidad numérica en modelos grandes.
#
# torch_dtype=bfloat16: alinea las capas no cuantizadas
#   (embedding, lm_head) con compute_dtype. Ver sección 0.
#
# device_map="auto": distribuye capas automáticamente en GPU.
#
# padding_side="left": requerido por la arquitectura de
#   Mistral para generación causal correcta. Ver sección 0.
# ============================================================

print("="*60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4")
print("="*60)

clear_cache()

nf4_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=nf4_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
)
model.eval()

n_params = sum(p.numel() for p in model.parameters())
print(f"Parámetros totales: {n_params:,}")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ============================================================
# 3. PROMPT
# ============================================================

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal y en un único fenómeno: "
    "el gerundio de posterioridad.\n\n"

    "TAREA:\n"
    "Corrige SOLO los casos en los que el gerundio expresa una acción posterior a la del verbo principal.\n\n"

    "DEFINICIÓN:\n"
    "Hay gerundio de posterioridad únicamente cuando la acción expresada por el gerundio ocurre después de la acción del verbo principal. "
    "Si el gerundio expresa simultaneidad, modo, medio, causa, condición, descripción o forma parte de una perífrasis, no debe corregirse.\n\n"

    "OBJETIVO:\n"
    "Transforma esos casos en una coordinación con 'y' y un verbo conjugado.\n\n"

    "REGLA PRINCIPAL:\n"
    "Si hay duda sobre si el gerundio expresa posterioridad real, NO corrijas.\n\n"

    "CORREGIR SOLO SI:\n"
    "- el gerundio expresa con claridad una acción posterior;\n"
    "- la acción del gerundio ocurre después de la acción del verbo principal;\n"
    "- la corrección puede hacerse sustituyendo el gerundio por 'y' + verbo conjugado;\n"
    "- el nuevo verbo puede mantenerse en el mismo tiempo y modo que el verbo principal sin reformular libremente la oración.\n\n"

    "CÓMO CORREGIR:\n"
    "- Sustituye solo el segmento en gerundio afectado.\n"
    "- Usa 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo sujeto.\n"
    "- Ajusta pronombres o concordancia solo si es imprescindible para que la oración sea gramatical.\n"
    "- Conserva el significado original.\n\n"

    "NO CORREGIR NUNCA SI:\n"
    "- el gerundio expresa simultaneidad con el verbo principal;\n"
    "- el gerundio expresa modo o manera: 'habló sonriendo', 'respondió mirando al suelo';\n"
    "- el gerundio expresa medio o instrumento;\n"
    "- el gerundio expresa causa, condición o descripción;\n"
    "- el gerundio forma parte de una perífrasis o construcción aspectual: 'está estudiando', 'sigue trabajando', 'continúa avanzando', 'acabó aceptando', 'terminó cediendo', 'lleva años investigando';\n"
    "- el gerundio aparece dentro de comillas, cita directa o estilo directo;\n"
    "- la frase no contiene gerundio;\n"
    "- la frase ya es correcta y no contiene gerundio de posterioridad;\n"
    "- la modificación exigiría reescribir libremente la oración o alterar partes ajenas al fenómeno.\n\n"

    "RESTRICCIONES DE SALIDA:\n"
    "- Devuelve una sola oración final.\n"
    "- No expliques nada.\n"
    "- No añadas comentarios.\n"
    "- No cambies palabras no relacionadas con el fenómeno.\n"
    "- No cambies léxico, nombres, números, puntuación, mayúsculas, comillas ni tildes, salvo lo imprescindible para la corrección mínima.\n"
    "- No resumas ni reformules.\n"
    "- Conserva exactamente el contenido original excepto los cambios mínimos necesarios para sustituir el gerundio de posterioridad.\n\n"

    "CRITERIO DE CONSERVACIÓN:\n"
    "Si solo una parte de la oración contiene un gerundio de posterioridad, corrige solo esa parte y deja intacto todo lo demás.\n\n"

    "IMPORTANTE:\n"
    "Es mejor dejar una frase sin corregir que hacer una corrección falsa."
)

USER_PROMPT = (
    "Ejemplos:\n\n"

    "Input: El laboratorio completó la primera fase del ensayo, publicando los resultados dos meses después.\n"
    "Output: El laboratorio completó la primera fase del ensayo y publicó los resultados dos meses después.\n\n"

    "Input: La comisaria inaugurará la exposición temporal, trasladándola después a la sede principal del museo.\n"
    "Output: La comisaria inaugurará la exposición temporal y la trasladará después a la sede principal del museo.\n\n"

    "Input: El decano presentó el nuevo plan docente, remitiéndolo al claustro al día siguiente para su valoración.\n"
    "Output: El decano presentó el nuevo plan docente y lo remitió al claustro al día siguiente para su valoración.\n\n"

    "Input: La empresa validó el prototipo final, registrando la patente semanas más tarde.\n"
    "Output: La empresa validó el prototipo final y registró la patente semanas más tarde.\n\n"

    "Input: El bibliotecario ordenará los manuscritos antiguos, digitalizándolos después para el archivo central.\n"
    "Output: El bibliotecario ordenará los manuscritos antiguos y los digitalizará después para el archivo central.\n\n"

    "Input: La arquitecta restauró la fachada principal, reabriendo el edificio meses después al público general.\n"
    "Output: La arquitecta restauró la fachada principal y reabrió el edificio meses después al público general.\n\n"

    "Input: El rector compareció ante la prensa, anunciando la decisión definitiva horas más tarde en un comunicado oficial.\n"
    "Output: El rector compareció ante la prensa y anunció la decisión definitiva horas más tarde en un comunicado oficial.\n\n"

    "Input: El equipo entregó la versión preliminar del informe, incorporando las correcciones la semana siguiente.\n"
    "Output: El equipo entregó la versión preliminar del informe e incorporó las correcciones la semana siguiente.\n\n"

    "Input: La profesora explicó el tema utilizando ejemplos del examen anterior.\n"
    "Output: La profesora explicó el tema utilizando ejemplos del examen anterior.\n\n"

    "Input: El violinista interpretó la pieza mirando al director en los compases finales.\n"
    "Output: El violinista interpretó la pieza mirando al director en los compases finales.\n\n"

    "Input: El departamento sigue revisando las solicitudes pendientes.\n"
    "Output: El departamento sigue revisando las solicitudes pendientes.\n\n"

    "Input: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n"
    "Output: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n\n"

    "Input: \"Estoy preparando la defensa\", respondió la doctoranda.\n"
    "Output: \"Estoy preparando la defensa\", respondió la doctoranda.\n\n"

    "Input: El senderista avanzaba observando el paisaje desde la ladera.\n"
    "Output: El senderista avanzaba observando el paisaje desde la ladera.\n\n"

    "Input: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n"
    "Output: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

def build_prompt(input_text: str):
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT.format(sentence=input_text)},
    ]

# ============================================================
# 4. DATASET Y DATALOADER
# ============================================================

class GECDatasetMistral(Dataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return self.inputs[idx], self.targets[idx]

# batch_size=1: obligatorio para LLMs causales en inferencia.
test_loader = DataLoader(
    GECDatasetMistral(df_test),
    batch_size=1, shuffle=False,
)
noticias_loader = DataLoader(
    GECDatasetMistral(df_noticias_final),
    batch_size=1, shuffle=False,
)

print(f"Test: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")

# ============================================================
# 5. MÉTRICAS GEC
#
# Sistema de métricas idéntico al de BiLSTM, mBART, mT5 y
# Llama para garantizar comparabilidad directa en la tabla
# del paper.
#
# NIVEL 1 — Detección (clasificación binaria a nivel de oración):
#   Positivo  = src != tgt → la frase tiene errores
#   Negativo  = src == tgt → la frase es correcta
#   Pred pos  = pred != src → el modelo modifica
#   Pred neg  = pred == src → el modelo no toca nada
#
#   TP: tenía errores  + modelo modifica    ✓
#   TN: era correcta   + modelo no toca     ✓
#   FP: era correcta   + modelo modifica    ✗ falsa corrección
#   FN: tenía errores  + modelo no toca     ✗ no corrige
#
#   Accuracy  = (TP + TN) / N
#   Precision = TP / (TP + FP)
#   Recall    = TP / (TP + FN)
#   F0.5      = penaliza más FP que FN (estándar GEC)
#
# NIVEL 2 — Calidad de corrección (token-level):
#   GLEU, F0.5 tok, precision tok, recall tok, ED norm.
#   Calculadas sobre palabras (split por espacio), igual que
#   el resto de modelos. No sobre subpalabras de Mistral.
# ============================================================

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids, skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()

def clean_pred(text: str) -> str:
    """
    Limpia la salida de Mistral.
    Mistral Instruct puede añadir prefijos o comillas incluso
    con la instrucción explícita de no hacerlo. Se amplía
    respecto a Llama con 'output' y 'respuesta' porque el
    chat template de Mistral puede introducirlos. La lógica
    de extracción de primera oración es idéntica a Llama.
    """
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    # Nivel 1: detección binaria
    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()

    # Nivel 2: calidad token-level
    gleu = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )

    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}

    n = len(all_metrics)

    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05 = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )

    return {
        "n_total":          n,
        "n_positive":       tp + fn,
        "n_negative":       tn + fp,
        "tp":               tp,
        "tn":               tn,
        "fp":               fp,
        "fn":               fn,
        "accuracy":         accuracy,
        "precision":        precision,
        "recall":           recall,
        "f05":              f05,
        "gleu_mean":        float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":         float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":     float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":      float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":    float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":       float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":     float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":  float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS — {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc = p["needs_correction"]
        pm = p["pred_modified"]
        if     nc and     pm: label = "✓ TP  (con errores → modelo corrige)"
        elif not nc and not pm: label = "✓ TN  (sin errores → modelo no toca)"
        elif not nc and     pm: label = "✗ FP  (sin errores → modelo modifica igualmente)"
        else:                   label = "✗ FN  (con errores → modelo no corrige)"

        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(
            f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} "
            f"| ED_norm={p['ed_norm']:.3f}"
        )


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"MÉTRICAS AGREGADAS — {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  ── Detección (clasificación binaria) ──────────────")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  "
          f"(frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  "
          f"(de las que corrige, cuántas realmente tenían error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  "
          f"(de las que tenían error, cuántas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  "
          f"(penaliza más FP que FN, estándar GEC)")
    print()
    print(f"  ── Calidad de corrección (token-level) ────────────")
    print(f"     GLEU:           {agg.get('gleu_mean',     0):.4f} "
          f"± {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean',  0):.4f} "
          f"± {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  ── Comportamiento del modelo ───────────────────────")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  ── Tiempos ─────────────────────────────────────────")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ============================================================
# 6. FUNCIÓN DE INFERENCIA
# ============================================================

def evaluate_mistral(model, loader, src_texts_list=None):
    model.eval()
    all_metrics = []
    predictions = []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            messages = build_prompt(input_text)

            enc = tokenizer.apply_chat_template(
                messages,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids,
                attention_mask=attention_mask,
                max_new_tokens=MAX_NEW_TOKENS,
                do_sample=DO_SAMPLE,
                num_beams=NUM_BEAMS,
                repetition_penalty=REPETITION_PENALTY,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            # Extraer solo los tokens generados (no el prompt)
            generated_ids = outputs[0][input_length:]
            pred_raw  = decode_pred(generated_ids)
            pred_text = clean_pred(pred_raw)

            src_text = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src":      src_text,
                "tgt":      tgt_text,
                "pred_raw": pred_raw,
                "pred":     pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions

# ============================================================
# 7. EVALUACIÓN TEST
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN TEST SET — Mistral-7B-Instruct-v0.3 (Prompt 4 (NIVEL 4 — Few-shot + reglas))")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_TEST_START = time.perf_counter()

test_agg, test_preds = evaluate_mistral(
    model, test_loader,
    src_texts_list=df_test["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - EVAL_TEST_START

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time,
                         len(test_preds), "TEST SET")

# ============================================================
# 8. EVALUACIÓN NOTICIAS
# ============================================================

print("\n" + "="*60)
print("EVALUACIÓN EN NOTICIAS — Mistral-7B-Instruct-v0.3")
print("="*60)

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

if torch.cuda.is_available():
    torch.cuda.synchronize()
EVAL_NOT_START = time.perf_counter()

noticias_agg, noticias_preds = evaluate_mistral(
    model, noticias_loader,
    src_texts_list=df_noticias_final["input"].tolist(),
)

if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - EVAL_NOT_START

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time,
                         len(noticias_preds), "NOTICIAS")

# ============================================================
# 9. RESULTADOS FINALES
# ============================================================

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":        "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":   "Prompt 4 (NIVEL 4 — Few-shot + reglas)",
    "seed":              SEED,
    "device":            device,
    "n_params":          n_params,
    "quantization":      "nf4_4bit",
    "max_new_tokens":    MAX_NEW_TOKENS,
    "do_sample":         DO_SAMPLE,
    "num_beams":         NUM_BEAMS,
    "repetition_penalty": REPETITION_PENALTY,
    "decode_strategy":   "greedy",

    # Sin entrenamiento
    "total_train_time_s": 0.0,
    "epochs_run":         0,

    # TEST — clasificación binaria
    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    # TEST — token-level
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    # NOTICIAS — clasificación binaria
    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    # NOTICIAS — token-level
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "="*60)
print("RESULTADOS FINALES — RESUMEN")
print("="*60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral_zeroshot_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral_gerundio.json'")
```

**FINE-TUNING**

```python
# ===============================================================
# PIPELINE FINE-TUNING GEC — Mistral-7B-Instruct-v0.3 + QLoRA
# Unsloth 2026 — patrón oficial verificado
# ===============================================================
#
# INSTALACIÓN (ejecutar en celda separada antes de este script):
#   !pip install --upgrade "unsloth[colab-new]"
#
# DIFERENCIAS RESPECTO AL FINE-TUNING DE LLAMA:
#   - Mistral no soporta role="system" nativamente en su chat
#     template: Unsloth lo ignora o lo fusiona de forma imprevisible.
#     Solución: system + user se concatenan en un único turno "user",
#     igual que recomienda la documentación oficial de Mistral.
#   - train_on_responses_only usa los delimitadores de Mistral:
#     instruction_part="[INST] ", response_part="[/INST]"
#     (sin espacio tras [/INST], que es como genera el template).
#   - EOS token: Mistral usa </s> (id=2).
#   - padding_side="left" en inferencia (igual que en prompting).
#   - NO usar FastLanguageModel.for_inference(): misma razón que
#     en Llama (RuntimeError con prompts largos de few-shot).
# ===============================================================

# ---------------------------------------------------------------
# IMPORTANTE: unsloth SIEMPRE primero
# ---------------------------------------------------------------
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

import json, time, gc, re, random
import numpy as np
import torch
from torch.utils.data import Dataset as TorchDataset, DataLoader
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback
from datasets import Dataset as HFDataset

# ---------------------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------------------

SEED           = 42
MODEL_NAME     = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_SEQ_LENGTH = 2048
DTYPE          = None
LOAD_IN_4BIT   = True

LORA_R         = 8
LORA_ALPHA     = 32
LORA_DROPOUT   = 0.0

NUM_EPOCHS     = 2
LEARNING_RATE  = 2e-5
BATCH_SIZE     = 2
GRAD_ACCUM     = 4             # batch efectivo = 8
WARMUP_STEPS   = 20
PATIENCE       = 1

MAX_NEW_TOKENS     = 150
REPETITION_PENALTY = 1.00
MISTRAL_EOS_ID     = 2         # </s>

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ---------------------------------------------------------------
# CARGA DEL MODELO
# ---------------------------------------------------------------

print("\n" + "=" * 60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4 (QLoRA)")
print("=" * 60)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = DTYPE,
    load_in_4bit   = LOAD_IN_4BIT,
)

model = FastLanguageModel.get_peft_model(
    model,
    r              = LORA_R,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha     = LORA_ALPHA,
    lora_dropout   = LORA_DROPOUT,
    bias           = "none",
    use_gradient_checkpointing = "unsloth",
    random_state   = SEED,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "right"

n_params       = sum(p.numel() for p in model.parameters())
n_params_train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parametros totales:      {n_params:,}")
print(f"Parametros entrenables:  {n_params_train:,} ({100*n_params_train/n_params:.2f}%)")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PROMPT
#
# Mistral no soporta role="system" en su chat template nativo.
# Se fusiona SYSTEM_PROMPT + USER_PROMPT en un único turno "user",
# que es el patrón correcto para Mistral y garantiza que
# train_on_responses_only encuentre correctamente el [/INST].
# ---------------------------------------------------------------

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal y en un único fenómeno: "
    "el gerundio de posterioridad.\n\n"

    "TAREA:\n"
    "Corrige SOLO los casos en los que el gerundio expresa una acción posterior a la del verbo principal.\n\n"

    "DEFINICIÓN:\n"
    "Hay gerundio de posterioridad únicamente cuando la acción expresada por el gerundio ocurre después de la acción del verbo principal. "
    "Si el gerundio expresa simultaneidad, anterioridad, modo, medio, causa, condición, descripción o forma parte de una perífrasis, no debe corregirse.\n\n"

    "OBJETIVO:\n"
    "Transforma esos casos en una coordinación con 'y' y un verbo conjugado.\n\n"

    "REGLA PRINCIPAL:\n"
    "Si hay duda sobre si el gerundio expresa posterioridad real, NO corrijas.\n\n"

    "CORREGIR SOLO SI:\n"
    "- el gerundio expresa con claridad una acción posterior;\n"
    "- la acción del gerundio ocurre después de la acción del verbo principal;\n"
    "- la corrección puede hacerse sustituyendo el gerundio por 'y' + verbo conjugado;\n"
    "- el nuevo verbo puede mantenerse en el mismo tiempo y modo que el verbo principal sin reformular libremente la oración.\n\n"

    "CÓMO CORREGIR:\n"
    "- Sustituye solo el segmento en gerundio afectado.\n"
    "- Usa 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo sujeto.\n"
    "- Ajusta pronombres o concordancia solo si es imprescindible para que la oración sea gramatical.\n"
    "- Conserva el significado original.\n\n"

    "NO CORREGIR NUNCA SI:\n"
    "- el gerundio expresa simultaneidad con el verbo principal;\n"
    "- el gerundio expresa anterioridad;\n"
    "- el gerundio expresa modo o manera: 'habló sonriendo', 'respondió mirando al suelo';\n"
    "- el gerundio expresa medio o instrumento;\n"
    "- el gerundio expresa causa, condición o descripción;\n"
    "- el gerundio forma parte de una perífrasis o construcción aspectual: 'está estudiando', 'sigue trabajando', 'continúa avanzando', 'acabó aceptando', 'terminó cediendo', 'lleva años investigando';\n"
    "- el gerundio aparece dentro de comillas, cita directa o estilo directo;\n"
    "- la frase no contiene gerundio;\n"
    "- la frase ya es correcta y no contiene gerundio de posterioridad;\n"
    "- la modificación exigiría reescribir libremente la oración o alterar partes ajenas al fenómeno.\n\n"

    "RESTRICCIONES DE SALIDA:\n"
    "- Devuelve una sola oración final.\n"
    "- No expliques nada.\n"
    "- No añadas comentarios.\n"
    "- No cambies palabras no relacionadas con el fenómeno.\n"
    "- No cambies léxico, nombres, números, puntuación, mayúsculas, comillas ni tildes, salvo lo imprescindible para la corrección mínima.\n"
    "- No resumas ni reformules.\n"
    "- Conserva exactamente el contenido original excepto los cambios mínimos necesarios para sustituir el gerundio de posterioridad.\n\n"

    "CRITERIO DE CONSERVACIÓN:\n"
    "Si solo una parte de la oración contiene un gerundio de posterioridad, corrige solo esa parte y deja intacto todo lo demás.\n\n"

    "IMPORTANTE:\n"
    "Es mejor dejar una frase sin corregir que hacer una corrección falsa."
)

USER_PROMPT = (
    "Ejemplos:\n\n"

    "Input: El laboratorio completó la primera fase del ensayo, publicando los resultados dos meses después.\n"
    "Output: El laboratorio completó la primera fase del ensayo y publicó los resultados dos meses después.\n\n"

    "Input: La comisaria inaugurará la exposición temporal, trasladándola después a la sede principal del museo.\n"
    "Output: La comisaria inaugurará la exposición temporal y la trasladará después a la sede principal del museo.\n\n"

    "Input: El decano presentó el nuevo plan docente, remitiéndolo al claustro al día siguiente para su valoración.\n"
    "Output: El decano presentó el nuevo plan docente y lo remitió al claustro al día siguiente para su valoración.\n\n"

    "Input: La empresa validó el prototipo final, registrando la patente semanas más tarde.\n"
    "Output: La empresa validó el prototipo final y registró la patente semanas más tarde.\n\n"

    "Input: El bibliotecario ordenará los manuscritos antiguos, digitalizándolos después para el archivo central.\n"
    "Output: El bibliotecario ordenará los manuscritos antiguos y los digitalizará después para el archivo central.\n\n"

    "Input: La arquitecta restauró la fachada principal, reabriendo el edificio meses después al público general.\n"
    "Output: La arquitecta restauró la fachada principal y reabrió el edificio meses después al público general.\n\n"

    "Input: El rector compareció ante la prensa, anunciando la decisión definitiva horas más tarde en un comunicado oficial.\n"
    "Output: El rector compareció ante la prensa y anunció la decisión definitiva horas más tarde en un comunicado oficial.\n\n"

    "Input: El equipo entregó la versión preliminar del informe, incorporando las correcciones la semana siguiente.\n"
    "Output: El equipo entregó la versión preliminar del informe e incorporó las correcciones la semana siguiente.\n\n"

    "Input: La profesora explicó el tema utilizando ejemplos del examen anterior.\n"
    "Output: La profesora explicó el tema utilizando ejemplos del examen anterior.\n\n"

    "Input: El violinista interpretó la pieza mirando al director en los compases finales.\n"
    "Output: El violinista interpretó la pieza mirando al director en los compases finales.\n\n"

    "Input: El departamento sigue revisando las solicitudes pendientes.\n"
    "Output: El departamento sigue revisando las solicitudes pendientes.\n\n"

    "Input: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n"
    "Output: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n\n"

    "Input: \"Estoy preparando la defensa\", respondió la doctoranda.\n"
    "Output: \"Estoy preparando la defensa\", respondió la doctoranda.\n\n"

    "Input: El senderista avanzaba observando el paisaje desde la ladera.\n"
    "Output: El senderista avanzaba observando el paisaje desde la ladera.\n\n"

    "Input: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n"
    "Output: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

# System + user fusionados en un único turno "user" (patrón correcto para Mistral)
FULL_USER_PROMPT = SYSTEM_PROMPT + "\n\n" + USER_PROMPT

EOS_TOKEN = tokenizer.eos_token

def build_prompt(input_text: str) -> list[dict]:
    """Prompt de inferencia: system+user fusionados en un único turno user."""
    return [
        {"role": "user", "content": FULL_USER_PROMPT.format(sentence=input_text)},
    ]

def format_for_training(input_text: str, target_text: str) -> str:
    messages = [
        {"role": "user",      "content": FULL_USER_PROMPT.format(sentence=input_text)},
        {"role": "assistant", "content": target_text},
    ]
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    ) + EOS_TOKEN

# ---------------------------------------------------------------
# DIAGNÓSTICO DE TOKENS
# ---------------------------------------------------------------

def diagnostico_tokens(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        text = (
            format_for_training(row["input"], row["target"])
            if include_target
            else tokenizer.apply_chat_template(
                build_prompt(row["input"]),
                tokenize=False,
                add_generation_prompt=True,
            )
        )
        lengths.append(len(tokenizer(text, add_special_tokens=False)["input_ids"]))
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_SEQ_LENGTH).sum()
    print(f"\n--- Tokens: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_SEQ_LENGTH={MAX_SEQ_LENGTH}: "
          f"{truncated}/{len(lengths)}", end="")
    print(
        f"  ATENCION: sube MAX_SEQ_LENGTH a {int(lengths.max()) + 64}"
        if truncated > 0 else "  OK"
    )
    return lengths

print("\n" + "=" * 60)
print("DIAGNOSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens(df_train,          "TRAIN    (prompt + target)", include_target=True)
diagnostico_tokens(df_val,            "VAL      (prompt + target)", include_target=True)
diagnostico_tokens(df_test,           "TEST     (solo prompt)",     include_target=False)
diagnostico_tokens(df_noticias_final, "NOTICIAS (solo prompt)",     include_target=False)

# ---------------------------------------------------------------
# DATASET
# ---------------------------------------------------------------

def make_hf_dataset(df):
    return HFDataset.from_dict({
        "text": [
            format_for_training(row["input"], row["target"])
            for _, row in df.iterrows()
        ]
    })

train_dataset = make_hf_dataset(df_train)
val_dataset   = make_hf_dataset(df_val)

print(f"\nTrain: {len(train_dataset)} | Val: {len(val_dataset)}")
print(f"Batch efectivo: {BATCH_SIZE} x {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ---------------------------------------------------------------
# ENTRENAMIENTO
#
# train_on_responses_only con los delimitadores correctos de Mistral:
#   instruction_part = "[INST] "   (inicio del turno user)
#   response_part    = "[/INST]"   (inicio del turno assistant, SIN espacio)
# El template de Mistral genera: <s>[INST] {user} [/INST]{target}</s>
# ---------------------------------------------------------------

trainer = SFTTrainer(
    model              = model,
    tokenizer          = tokenizer,
    train_dataset      = train_dataset,
    eval_dataset       = val_dataset,
    dataset_text_field = "text",
    max_seq_length     = MAX_SEQ_LENGTH,
    dataset_num_proc   = 2,
    packing            = False,
    args = SFTConfig(
        output_dir                  = "mistral7b_gerundio_lora",
        per_device_train_batch_size = BATCH_SIZE,
        gradient_accumulation_steps = GRAD_ACCUM,
        num_train_epochs            = NUM_EPOCHS,
        learning_rate               = LEARNING_RATE,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.05,
        neftune_noise_alpha         = 5,
        fp16                        = not is_bfloat16_supported(),
        bf16                        = is_bfloat16_supported(),
        logging_steps               = 5,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        greater_is_better           = False,
        optim                       = "paged_adamw_8bit",
        seed                        = SEED,
        report_to                   = "none",
    ),
    callbacks = [EarlyStoppingCallback(early_stopping_patience=PATIENCE)],
)

# Delimitadores correctos para Mistral-7B-Instruct-v0.3
# Template: <s>[INST] {user} [/INST]{assistant}</s>
trainer = train_on_responses_only(
    trainer,
    instruction_part = "[INST] ",
    response_part    = "[/INST]",
)

print("\nEmpieza entrenamiento Mistral-7B-Instruct-v0.3 (QLoRA)")
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()
TRAIN_START = time.perf_counter()
trainer.train()
total_train_time = time.perf_counter() - TRAIN_START
print(f"Entrenamiento completado en {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
if device == "cuda":
    print(f"Pico de memoria: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PREPARAR PARA INFERENCIA
# ---------------------------------------------------------------

model.eval()
tokenizer.padding_side = "left"

# ---------------------------------------------------------------
# MÉTRICAS GEC — idénticas al prompting de Mistral
# ---------------------------------------------------------------

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()


def clean_pred(text: str) -> str:
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    if not pred_text:
        pred_text = src_text

    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()
    gleu             = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta    = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )
    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}
    n  = len(all_metrics)
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05       = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )
    return {
        "n_total":           n,
        "n_positive":        tp + fn,
        "n_negative":        tn + fp,
        "tp": tp, "tn": tn, "fp": fp, "fn": fn,
        "accuracy":          accuracy,
        "precision":         precision,
        "recall":            recall,
        "f05":               f05,
        "gleu_mean":         float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":          float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":      float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":       float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":     float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":        float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":      float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":   float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS -- {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc, pm = p["needs_correction"], p["pred_modified"]
        if   nc and     pm: label = "TP  (con errores -> modelo corrige)"
        elif nc and not pm: label = "FN  (con errores -> modelo no corrige)"
        elif not nc and pm: label = "FP  (sin errores -> modelo modifica igualmente)"
        else:               label = "TN  (sin errores -> modelo no toca)"
        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} | ED_norm={p['ed_norm']:.3f}")


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"METRICAS AGREGADAS -- {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  -- Deteccion (clasificacion binaria) --")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  (frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  (de las que corrige, cuantas tenian error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  (de las que tenian error, cuantas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  (penaliza mas FP que FN, estandar GEC)")
    print()
    print(f"  -- Calidad de correccion (token-level) --")
    print(f"     GLEU:           {agg.get('gleu_mean',    0):.4f} +/- {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean', 0):.4f} +/- {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  -- Comportamiento del modelo --")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  -- Tiempos --")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ---------------------------------------------------------------
# EVALUACIÓN
# ---------------------------------------------------------------

class GECDataset(TorchDataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()
    def __len__(self):  return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

test_loader     = DataLoader(GECDataset(df_test),           batch_size=1, shuffle=False)
noticias_loader = DataLoader(GECDataset(df_noticias_final), batch_size=1, shuffle=False)
print(f"\nTest: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")


def evaluate_model(loader, src_texts_list=None):
    model.eval()
    all_metrics, predictions = [], []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            enc = tokenizer.apply_chat_template(
                build_prompt(input_text),
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids          = input_ids,
                attention_mask     = attention_mask,
                max_new_tokens     = MAX_NEW_TOKENS,
                do_sample          = False,
                num_beams          = 1,
                repetition_penalty = REPETITION_PENALTY,
                use_cache          = False,
                pad_token_id       = tokenizer.pad_token_id,
                eos_token_id       = MISTRAL_EOS_ID,
            )

            pred_raw  = decode_pred(outputs[0][input_length:])
            pred_text = clean_pred(pred_raw)
            src_text  = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src": src_text, "tgt": tgt_text,
                "pred_raw": pred_raw, "pred": pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions


# TEST
print("\n" + "=" * 60)
print("EVALUACION EN TEST SET — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

t0 = time.perf_counter()
test_agg, test_preds = evaluate_model(test_loader, df_test["input"].tolist())
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - t0

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time, len(test_preds), "TEST SET")

# NOTICIAS
print("\n" + "=" * 60)
print("EVALUACION EN NOTICIAS — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

t0 = time.perf_counter()
noticias_agg, noticias_preds = evaluate_model(
    noticias_loader, df_noticias_final["input"].tolist()
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - t0

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time, len(noticias_preds), "NOTICIAS")

# ---------------------------------------------------------------
# RESULTADOS FINALES
# ---------------------------------------------------------------

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":          "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":     "fine-tuned QLoRA",
    "seed":                SEED,
    "device":              device,
    "n_params":            n_params,
    "n_params_trainable":  n_params_train,
    "quantization":        "nf4_4bit",
    "lora_r":              LORA_R,
    "lora_alpha":          LORA_ALPHA,
    "lora_dropout":        LORA_DROPOUT,
    "epochs":              NUM_EPOCHS,
    "learning_rate":       LEARNING_RATE,
    "warmup_steps":        WARMUP_STEPS,
    "patience":            PATIENCE,
    "neftune_noise_alpha": 5,
    "batch_size":          BATCH_SIZE,
    "grad_accum":          GRAD_ACCUM,
    "effective_batch":     BATCH_SIZE * GRAD_ACCUM,
    "max_seq_length":      MAX_SEQ_LENGTH,
    "max_new_tokens":      MAX_NEW_TOKENS,
    "do_sample":           False,
    "repetition_penalty":  REPETITION_PENALTY,
    "decode_strategy":     "greedy",
    "loss_masking":        "train_on_responses_only",
    "total_train_time_s":  total_train_time,

    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES -- RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral7b_finetuned_gerundio.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral7b_finetuned_gerundio.json'")
```

```python
# ===============================================================
# PIPELINE FINE-TUNING GEC — Mistral-7B-Instruct-v0.3 + QLoRA
# Unsloth 2026 — patrón oficial verificado
# ===============================================================
#
# INSTALACIÓN (ejecutar en celda separada antes de este script):
#   !pip install --upgrade "unsloth[colab-new]"
#
# DIFERENCIAS RESPECTO AL FINE-TUNING DE LLAMA:
#   - Mistral no soporta role="system" nativamente en su chat
#     template: Unsloth lo ignora o lo fusiona de forma imprevisible.
#     Solución: system + user se concatenan en un único turno "user",
#     igual que recomienda la documentación oficial de Mistral.
#   - train_on_responses_only usa los delimitadores de Mistral:
#     instruction_part="[INST] ", response_part="[/INST]"
#     (sin espacio tras [/INST], que es como genera el template).
#   - EOS token: Mistral usa </s> (id=2).
#   - padding_side="left" en inferencia (igual que en prompting).
#   - NO usar FastLanguageModel.for_inference(): misma razón que
#     en Llama (RuntimeError con prompts largos de few-shot).
# ===============================================================

# ---------------------------------------------------------------
# IMPORTANTE: unsloth SIEMPRE primero
# ---------------------------------------------------------------
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

import json, time, gc, re, random
import numpy as np
import torch
from torch.utils.data import Dataset as TorchDataset, DataLoader
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback
from datasets import Dataset as HFDataset

# ---------------------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------------------

SEED           = 42
MODEL_NAME     = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_SEQ_LENGTH = 2048
DTYPE          = None
LOAD_IN_4BIT   = True

LORA_R         = 8
LORA_ALPHA     = 32
LORA_DROPOUT   = 0.0

NUM_EPOCHS     = 2
LEARNING_RATE  = 2e-5
BATCH_SIZE     = 2
GRAD_ACCUM     = 4             # batch efectivo = 8
WARMUP_STEPS   = 20
PATIENCE       = 1

MAX_NEW_TOKENS     = 150
REPETITION_PENALTY = 1.00
MISTRAL_EOS_ID     = 2         # </s>

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ---------------------------------------------------------------
# CARGA DEL MODELO
# ---------------------------------------------------------------

print("\n" + "=" * 60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4 (QLoRA)")
print("=" * 60)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = DTYPE,
    load_in_4bit   = LOAD_IN_4BIT,
)

model = FastLanguageModel.get_peft_model(
    model,
    r              = LORA_R,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha     = LORA_ALPHA,
    lora_dropout   = LORA_DROPOUT,
    bias           = "none",
    use_gradient_checkpointing = "unsloth",
    random_state   = SEED,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "right"

n_params       = sum(p.numel() for p in model.parameters())
n_params_train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parametros totales:      {n_params:,}")
print(f"Parametros entrenables:  {n_params_train:,} ({100*n_params_train/n_params:.2f}%)")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PROMPT
#
# Mistral no soporta role="system" en su chat template nativo.
# Se fusiona SYSTEM_PROMPT + USER_PROMPT en un único turno "user",
# que es el patrón correcto para Mistral y garantiza que
# train_on_responses_only encuentre correctamente el [/INST].
#
# El prompt de inferencia (build_prompt) usa el mismo formato,
# manteniendo coherencia con el pipeline de prompting (doc 3).
# ---------------------------------------------------------------

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal y en un único fenómeno: "
    "el gerundio de posterioridad.\n\n"

    "TAREA:\n"
    "Corrige SOLO los casos en los que el gerundio expresa una acción posterior a la del verbo principal.\n\n"

    "DEFINICIÓN:\n"
    "Hay gerundio de posterioridad únicamente cuando la acción expresada por el gerundio ocurre después de la acción del verbo principal. "
    "Si el gerundio expresa simultaneidad, modo, medio, causa, condición, descripción o forma parte de una perífrasis, no debe corregirse.\n\n"

    "OBJETIVO:\n"
    "Transforma esos casos en una coordinación con 'y' y un verbo conjugado.\n\n"

    "REGLA PRINCIPAL:\n"
    "Si hay duda sobre si el gerundio expresa posterioridad real, NO corrijas.\n\n"

    "CORREGIR SOLO SI:\n"
    "- el gerundio expresa con claridad una acción posterior;\n"
    "- la acción del gerundio ocurre después de la acción del verbo principal;\n"
    "- la corrección puede hacerse sustituyendo el gerundio por 'y' + verbo conjugado;\n"
    "- el nuevo verbo puede mantenerse en el mismo tiempo y modo que el verbo principal sin reformular libremente la oración.\n\n"

    "CÓMO CORREGIR:\n"
    "- Sustituye solo el segmento en gerundio afectado.\n"
    "- Usa 'y' + verbo conjugado.\n"
    "- Mantén el mismo tiempo verbal y modo que el verbo principal.\n"
    "- Mantén el mismo sujeto.\n"
    "- Ajusta pronombres o concordancia solo si es imprescindible para que la oración sea gramatical.\n"
    "- Conserva el significado original.\n\n"

    "NO CORREGIR NUNCA SI:\n"
    "- el gerundio expresa simultaneidad con el verbo principal;\n"
    "- el gerundio expresa modo o manera: 'habló sonriendo', 'respondió mirando al suelo';\n"
    "- el gerundio expresa medio o instrumento;\n"
    "- el gerundio expresa causa, condición o descripción;\n"
    "- el gerundio forma parte de una perífrasis o construcción aspectual: 'está estudiando', 'sigue trabajando', 'continúa avanzando', 'acabó aceptando', 'terminó cediendo', 'lleva años investigando';\n"
    "- el gerundio aparece dentro de comillas, cita directa o estilo directo;\n"
    "- la frase no contiene gerundio;\n"
    "- la frase ya es correcta y no contiene gerundio de posterioridad;\n"
    "- la modificación exigiría reescribir libremente la oración o alterar partes ajenas al fenómeno.\n\n"

    "RESTRICCIONES DE SALIDA:\n"
    "- Devuelve una sola oración final.\n"
    "- No expliques nada.\n"
    "- No añadas comentarios.\n"
    "- No cambies palabras no relacionadas con el fenómeno.\n"
    "- No cambies léxico, nombres, números, puntuación, mayúsculas, comillas ni tildes, salvo lo imprescindible para la corrección mínima.\n"
    "- No resumas ni reformules.\n"
    "- Conserva exactamente el contenido original excepto los cambios mínimos necesarios para sustituir el gerundio de posterioridad.\n\n"

    "CRITERIO DE CONSERVACIÓN:\n"
    "Si solo una parte de la oración contiene un gerundio de posterioridad, corrige solo esa parte y deja intacto todo lo demás.\n\n"

    "IMPORTANTE:\n"
    "Es mejor dejar una frase sin corregir que hacer una corrección falsa."
)

USER_PROMPT = (
    "Ejemplos:\n\n"

    "Input: El laboratorio completó la primera fase del ensayo, publicando los resultados dos meses después.\n"
    "Output: El laboratorio completó la primera fase del ensayo y publicó los resultados dos meses después.\n\n"

    "Input: La comisaria inaugurará la exposición temporal, trasladándola después a la sede principal del museo.\n"
    "Output: La comisaria inaugurará la exposición temporal y la trasladará después a la sede principal del museo.\n\n"

    "Input: El decano presentó el nuevo plan docente, remitiéndolo al claustro al día siguiente para su valoración.\n"
    "Output: El decano presentó el nuevo plan docente y lo remitió al claustro al día siguiente para su valoración.\n\n"

    "Input: La empresa validó el prototipo final, registrando la patente semanas más tarde.\n"
    "Output: La empresa validó el prototipo final y registró la patente semanas más tarde.\n\n"

    "Input: El bibliotecario ordenará los manuscritos antiguos, digitalizándolos después para el archivo central.\n"
    "Output: El bibliotecario ordenará los manuscritos antiguos y los digitalizará después para el archivo central.\n\n"

    "Input: La arquitecta restauró la fachada principal, reabriendo el edificio meses después al público general.\n"
    "Output: La arquitecta restauró la fachada principal y reabrió el edificio meses después al público general.\n\n"

    "Input: El rector compareció ante la prensa, anunciando la decisión definitiva horas más tarde en un comunicado oficial.\n"
    "Output: El rector compareció ante la prensa y anunció la decisión definitiva horas más tarde en un comunicado oficial.\n\n"

    "Input: El equipo entregó la versión preliminar del informe, incorporando las correcciones la semana siguiente.\n"
    "Output: El equipo entregó la versión preliminar del informe e incorporó las correcciones la semana siguiente.\n\n"

    "Input: La profesora explicó el tema utilizando ejemplos del examen anterior.\n"
    "Output: La profesora explicó el tema utilizando ejemplos del examen anterior.\n\n"

    "Input: El violinista interpretó la pieza mirando al director en los compases finales.\n"
    "Output: El violinista interpretó la pieza mirando al director en los compases finales.\n\n"

    "Input: El departamento sigue revisando las solicitudes pendientes.\n"
    "Output: El departamento sigue revisando las solicitudes pendientes.\n\n"

    "Input: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n"
    "Output: La comisión terminó aceptando la propuesta tras varias rondas de negociación.\n\n"

    "Input: \"Estoy preparando la defensa\", respondió la doctoranda.\n"
    "Output: \"Estoy preparando la defensa\", respondió la doctoranda.\n\n"

    "Input: El senderista avanzaba observando el paisaje desde la ladera.\n"
    "Output: El senderista avanzaba observando el paisaje desde la ladera.\n\n"

    "Input: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n"
    "Output: En los pasillos del edificio, en el archivo o revisando las cajas antiguas.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

# System + user fusionados en un único turno "user" (patrón correcto para Mistral)
FULL_USER_PROMPT = SYSTEM_PROMPT + "\n\n" + USER_PROMPT

EOS_TOKEN = tokenizer.eos_token

def build_prompt(input_text: str) -> list[dict]:
    """Prompt de inferencia: system+user fusionados en un único turno user."""
    return [
        {"role": "user", "content": FULL_USER_PROMPT.format(sentence=input_text)},
    ]

def format_for_training(input_text: str, target_text: str) -> str:
    messages = [
        {"role": "user",      "content": FULL_USER_PROMPT.format(sentence=input_text)},
        {"role": "assistant", "content": target_text},
    ]
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    ) + EOS_TOKEN

# ---------------------------------------------------------------
# DIAGNÓSTICO DE TOKENS
# ---------------------------------------------------------------

def diagnostico_tokens(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        text = (
            format_for_training(row["input"], row["target"])
            if include_target
            else tokenizer.apply_chat_template(
                build_prompt(row["input"]),
                tokenize=False,
                add_generation_prompt=True,
            )
        )
        lengths.append(len(tokenizer(text, add_special_tokens=False)["input_ids"]))
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_SEQ_LENGTH).sum()
    print(f"\n--- Tokens: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_SEQ_LENGTH={MAX_SEQ_LENGTH}: "
          f"{truncated}/{len(lengths)}", end="")
    print(
        f"  ATENCION: sube MAX_SEQ_LENGTH a {int(lengths.max()) + 64}"
        if truncated > 0 else "  OK"
    )
    return lengths

print("\n" + "=" * 60)
print("DIAGNOSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens(df_train,          "TRAIN    (prompt + target)", include_target=True)
diagnostico_tokens(df_val,            "VAL      (prompt + target)", include_target=True)
diagnostico_tokens(df_test,           "TEST     (solo prompt)",     include_target=False)
diagnostico_tokens(df_noticias_final, "NOTICIAS (solo prompt)",     include_target=False)

# ---------------------------------------------------------------
# DATASET
# ---------------------------------------------------------------

def make_hf_dataset(df):
    return HFDataset.from_dict({
        "text": [
            format_for_training(row["input"], row["target"])
            for _, row in df.iterrows()
        ]
    })

train_dataset = make_hf_dataset(df_train)
val_dataset   = make_hf_dataset(df_val)

print(f"\nTrain: {len(train_dataset)} | Val: {len(val_dataset)}")
print(f"Batch efectivo: {BATCH_SIZE} x {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ---------------------------------------------------------------
# ENTRENAMIENTO
#
# train_on_responses_only con los delimitadores correctos de Mistral:
#   instruction_part = "[INST] "   (inicio del turno user)
#   response_part    = "[/INST]"   (inicio del turno assistant, SIN espacio)
# El template de Mistral genera: <s>[INST] {user} [/INST]{target}</s>
# ---------------------------------------------------------------

trainer = SFTTrainer(
    model              = model,
    tokenizer          = tokenizer,
    train_dataset      = train_dataset,
    eval_dataset       = val_dataset,
    dataset_text_field = "text",
    max_seq_length     = MAX_SEQ_LENGTH,
    dataset_num_proc   = 2,
    packing            = False,
    args = SFTConfig(
        output_dir                  = "mistral7b_tu_lora",
        per_device_train_batch_size = BATCH_SIZE,
        gradient_accumulation_steps = GRAD_ACCUM,
        num_train_epochs            = NUM_EPOCHS,
        learning_rate               = LEARNING_RATE,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.05,
        neftune_noise_alpha         = 5,
        fp16                        = not is_bfloat16_supported(),
        bf16                        = is_bfloat16_supported(),
        logging_steps               = 5,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        greater_is_better           = False,
        optim                       = "paged_adamw_8bit",
        seed                        = SEED,
        report_to                   = "none",
    ),
    callbacks = [EarlyStoppingCallback(early_stopping_patience=PATIENCE)],
)

# Delimitadores correctos para Mistral-7B-Instruct-v0.3
# Template: <s>[INST] {user} [/INST]{assistant}</s>
trainer = train_on_responses_only(
    trainer,
    instruction_part = "[INST] ",
    response_part    = "[/INST]",
)

print("\nEmpieza entrenamiento Mistral-7B-Instruct-v0.3 (QLoRA)")
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()
TRAIN_START = time.perf_counter()
trainer.train()
total_train_time = time.perf_counter() - TRAIN_START
print(f"Entrenamiento completado en {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
if device == "cuda":
    print(f"Pico de memoria: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PREPARAR PARA INFERENCIA
# ---------------------------------------------------------------

model.eval()
tokenizer.padding_side = "left"

# ---------------------------------------------------------------
# MÉTRICAS GEC — idénticas al prompting de Mistral (doc 3)
# ---------------------------------------------------------------

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()


def clean_pred(text: str) -> str:
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    if not pred_text:
        pred_text = src_text

    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()
    gleu             = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta    = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )
    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}
    n  = len(all_metrics)
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05       = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )
    return {
        "n_total":           n,
        "n_positive":        tp + fn,
        "n_negative":        tn + fp,
        "tp": tp, "tn": tn, "fp": fp, "fn": fn,
        "accuracy":          accuracy,
        "precision":         precision,
        "recall":            recall,
        "f05":               f05,
        "gleu_mean":         float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":          float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":      float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":       float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":     float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":        float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":      float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":   float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS -- {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc, pm = p["needs_correction"], p["pred_modified"]
        if   nc and     pm: label = "TP  (con errores -> modelo corrige)"
        elif nc and not pm: label = "FN  (con errores -> modelo no corrige)"
        elif not nc and pm: label = "FP  (sin errores -> modelo modifica igualmente)"
        else:               label = "TN  (sin errores -> modelo no toca)"
        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} | ED_norm={p['ed_norm']:.3f}")


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"METRICAS AGREGADAS -- {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  -- Deteccion (clasificacion binaria) --")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  (frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  (de las que corrige, cuantas tenian error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  (de las que tenian error, cuantas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  (penaliza mas FP que FN, estandar GEC)")
    print()
    print(f"  -- Calidad de correccion (token-level) --")
    print(f"     GLEU:           {agg.get('gleu_mean',    0):.4f} +/- {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean', 0):.4f} +/- {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  -- Comportamiento del modelo --")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  -- Tiempos --")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ---------------------------------------------------------------
# EVALUACIÓN
# ---------------------------------------------------------------

class GECDataset(TorchDataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()
    def __len__(self):  return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

test_loader     = DataLoader(GECDataset(df_test),           batch_size=1, shuffle=False)
noticias_loader = DataLoader(GECDataset(df_noticias_final), batch_size=1, shuffle=False)
print(f"\nTest: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")


def evaluate_model(loader, src_texts_list=None):
    model.eval()
    all_metrics, predictions = [], []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            enc = tokenizer.apply_chat_template(
                build_prompt(input_text),
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids          = input_ids,
                attention_mask     = attention_mask,
                max_new_tokens     = MAX_NEW_TOKENS,
                do_sample          = False,
                num_beams          = 1,
                repetition_penalty = REPETITION_PENALTY,
                use_cache          = False,
                pad_token_id       = tokenizer.pad_token_id,
                eos_token_id       = MISTRAL_EOS_ID,
            )

            pred_raw  = decode_pred(outputs[0][input_length:])
            pred_text = clean_pred(pred_raw)
            src_text  = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src": src_text, "tgt": tgt_text,
                "pred_raw": pred_raw, "pred": pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions


# TEST
print("\n" + "=" * 60)
print("EVALUACION EN TEST SET — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

t0 = time.perf_counter()
test_agg, test_preds = evaluate_model(test_loader, df_test["input"].tolist())
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - t0

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time, len(test_preds), "TEST SET")

# NOTICIAS
print("\n" + "=" * 60)
print("EVALUACION EN NOTICIAS — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

t0 = time.perf_counter()
noticias_agg, noticias_preds = evaluate_model(
    noticias_loader, df_noticias_final["input"].tolist()
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - t0

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time, len(noticias_preds), "NOTICIAS")

# ---------------------------------------------------------------
# RESULTADOS FINALES
# ---------------------------------------------------------------

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":          "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":     "fine-tuned QLoRA",
    "seed":                SEED,
    "device":              device,
    "n_params":            n_params,
    "n_params_trainable":  n_params_train,
    "quantization":        "nf4_4bit",
    "lora_r":              LORA_R,
    "lora_alpha":          LORA_ALPHA,
    "lora_dropout":        LORA_DROPOUT,
    "epochs":              NUM_EPOCHS,
    "learning_rate":       LEARNING_RATE,
    "warmup_steps":        WARMUP_STEPS,
    "patience":            PATIENCE,
    "neftune_noise_alpha": 5,
    "batch_size":          BATCH_SIZE,
    "grad_accum":          GRAD_ACCUM,
    "effective_batch":     BATCH_SIZE * GRAD_ACCUM,
    "max_seq_length":      MAX_SEQ_LENGTH,
    "max_new_tokens":      MAX_NEW_TOKENS,
    "do_sample":           False,
    "repetition_penalty":  REPETITION_PENALTY,
    "decode_strategy":     "greedy",
    "loss_masking":        "train_on_responses_only",
    "total_train_time_s":  total_train_time,

    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES -- RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral7b_finetuned.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral7b_finetuned.json'")
```

```python
# ===============================================================
# PIPELINE FINE-TUNING GEC — Mistral-7B-Instruct-v0.3 + QLoRA
# Unsloth 2026 — patrón oficial verificado
# ===============================================================
#
# INSTALACIÓN (ejecutar en celda separada antes de este script):
#   !pip install --upgrade "unsloth[colab-new]"
#
# DIFERENCIAS RESPECTO AL FINE-TUNING DE LLAMA:
#   - Mistral no soporta role="system" nativamente en su chat
#     template: Unsloth lo ignora o lo fusiona de forma imprevisible.
#     Solución: system + user se concatenan en un único turno "user",
#     igual que recomienda la documentación oficial de Mistral.
#   - train_on_responses_only usa los delimitadores de Mistral:
#     instruction_part="[INST] ", response_part="[/INST]"
#     (sin espacio tras [/INST], que es como genera el template).
#   - EOS token: Mistral usa </s> (id=2).
#   - padding_side="left" en inferencia (igual que en prompting).
#   - NO usar FastLanguageModel.for_inference(): misma razón que
#     en Llama (RuntimeError con prompts largos de few-shot).
# ===============================================================

# ---------------------------------------------------------------
# IMPORTANTE: unsloth SIEMPRE primero
# ---------------------------------------------------------------
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

import json, time, gc, re, random
import numpy as np
import torch
from torch.utils.data import Dataset as TorchDataset, DataLoader
from nltk.translate.gleu_score import sentence_gleu
from nltk.metrics import edit_distance
from collections import Counter
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback
from datasets import Dataset as HFDataset

# ---------------------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------------------

SEED           = 42
MODEL_NAME     = "mistralai/Mistral-7B-Instruct-v0.3"
MAX_SEQ_LENGTH = 2048
DTYPE          = None
LOAD_IN_4BIT   = True

LORA_R         = 8
LORA_ALPHA     = 32
LORA_DROPOUT   = 0.0

NUM_EPOCHS     = 2
LEARNING_RATE  = 2e-5
BATCH_SIZE     = 2
GRAD_ACCUM     = 4             # batch efectivo = 8
WARMUP_STEPS   = 20
PATIENCE       = 1

MAX_NEW_TOKENS     = 150
REPETITION_PENALTY = 1.00
MISTRAL_EOS_ID     = 2         # </s>

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo: {device}")
if device == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  Memoria total: {torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB")

# ---------------------------------------------------------------
# CARGA DEL MODELO
# ---------------------------------------------------------------

print("\n" + "=" * 60)
print("CARGANDO MISTRAL-7B-INSTRUCT-v0.3 EN 4-BIT NF4 (QLoRA)")
print("=" * 60)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = DTYPE,
    load_in_4bit   = LOAD_IN_4BIT,
)

model = FastLanguageModel.get_peft_model(
    model,
    r              = LORA_R,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha     = LORA_ALPHA,
    lora_dropout   = LORA_DROPOUT,
    bias           = "none",
    use_gradient_checkpointing = "unsloth",
    random_state   = SEED,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "right"

n_params       = sum(p.numel() for p in model.parameters())
n_params_train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parametros totales:      {n_params:,}")
print(f"Parametros entrenables:  {n_params_train:,} ({100*n_params_train/n_params:.2f}%)")
if device == "cuda":
    print(f"Memoria tras carga: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PROMPT
#
# Mistral no soporta role="system" en su chat template nativo.
# Se fusiona SYSTEM_PROMPT + USER_PROMPT en un único turno "user",
# que es el patrón correcto para Mistral y garantiza que
# train_on_responses_only encuentre correctamente el [/INST].
#
# El prompt de inferencia (build_prompt) usa el mismo formato,
# manteniendo coherencia con el pipeline de prompting (doc 3).
# ---------------------------------------------------------------

SYSTEM_PROMPT = (
    "Eres un corrector de gramática española especializado en lengua escrita formal. "
    "Corrige únicamente el gerundio de posterioridad. "
    "Devuelve solo la oración final."
)

USER_PROMPT = (
    "Input: La investigadora cerró la fase de recogida de datos, redactando el informe semanas después.\n"
    "Output: La investigadora cerró la fase de recogida de datos y redactó el informe semanas después.\n\n"

    "Input: El técnico revisará el servidor principal, reiniciándolo más tarde para aplicar la configuración definitiva.\n"
    "Output: El técnico revisará el servidor principal y lo reiniciará más tarde para aplicar la configuración definitiva.\n\n"

    "Input: El museo restauró la pieza dañada, exhibiéndola meses después en una sala temporal.\n"
    "Output: El museo restauró la pieza dañada y la exhibió meses después en una sala temporal.\n\n"

    "Input: La abogada preparó la documentación, presentándola al día siguiente en el registro.\n"
    "Output: La abogada preparó la documentación y la presentó al día siguiente en el registro.\n\n"

    "Input: El alumnado siguió trabajando en silencio durante la prueba.\n"
    "Output: El alumnado siguió trabajando en silencio durante la prueba.\n\n"

    "Input: El conferenciante habló sonriendo al público durante toda la sesión.\n"
    "Output: El conferenciante habló sonriendo al público durante toda la sesión.\n\n"

    "Input: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n"
    "Output: \"Estoy pensando en cambiar de despacho\", comentó el profesor.\n\n"

    "Input: El equipo terminó aceptando la propuesta tras varias horas de debate.\n"
    "Output: El equipo terminó aceptando la propuesta tras varias horas de debate.\n\n"

    "Input: {sentence}\n"
    "Output:"
)

# System + user fusionados en un único turno "user" (patrón correcto para Mistral)
FULL_USER_PROMPT = SYSTEM_PROMPT + "\n\n" + USER_PROMPT

EOS_TOKEN = tokenizer.eos_token

def build_prompt(input_text: str) -> list[dict]:
    """Prompt de inferencia: system+user fusionados en un único turno user."""
    return [
        {"role": "user", "content": FULL_USER_PROMPT.format(sentence=input_text)},
    ]

def format_for_training(input_text: str, target_text: str) -> str:
    messages = [
        {"role": "user",      "content": FULL_USER_PROMPT.format(sentence=input_text)},
        {"role": "assistant", "content": target_text},
    ]
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    ) + EOS_TOKEN

# ---------------------------------------------------------------
# DIAGNÓSTICO DE TOKENS
# ---------------------------------------------------------------

def diagnostico_tokens(df, nombre, include_target=False):
    lengths = []
    for _, row in df.iterrows():
        text = (
            format_for_training(row["input"], row["target"])
            if include_target
            else tokenizer.apply_chat_template(
                build_prompt(row["input"]),
                tokenize=False,
                add_generation_prompt=True,
            )
        )
        lengths.append(len(tokenizer(text, add_special_tokens=False)["input_ids"]))
    lengths   = np.array(lengths)
    truncated = (lengths > MAX_SEQ_LENGTH).sum()
    print(f"\n--- Tokens: {nombre} ---")
    print(f"  min={lengths.min()} | max={lengths.max()} | "
          f"media={lengths.mean():.1f} | p99={np.percentile(lengths,99):.0f}")
    print(f"  Truncados con MAX_SEQ_LENGTH={MAX_SEQ_LENGTH}: "
          f"{truncated}/{len(lengths)}", end="")
    print(
        f"  ATENCION: sube MAX_SEQ_LENGTH a {int(lengths.max()) + 64}"
        if truncated > 0 else "  OK"
    )
    return lengths

print("\n" + "=" * 60)
print("DIAGNOSTICO DE LONGITUD DE TOKENS")
print("=" * 60)
diagnostico_tokens(df_train,          "TRAIN    (prompt + target)", include_target=True)
diagnostico_tokens(df_val,            "VAL      (prompt + target)", include_target=True)
diagnostico_tokens(df_test,           "TEST     (solo prompt)",     include_target=False)
diagnostico_tokens(df_noticias_final, "NOTICIAS (solo prompt)",     include_target=False)

# ---------------------------------------------------------------
# DATASET
# ---------------------------------------------------------------

def make_hf_dataset(df):
    return HFDataset.from_dict({
        "text": [
            format_for_training(row["input"], row["target"])
            for _, row in df.iterrows()
        ]
    })

train_dataset = make_hf_dataset(df_train)
val_dataset   = make_hf_dataset(df_val)

print(f"\nTrain: {len(train_dataset)} | Val: {len(val_dataset)}")
print(f"Batch efectivo: {BATCH_SIZE} x {GRAD_ACCUM} = {BATCH_SIZE * GRAD_ACCUM}")

# ---------------------------------------------------------------
# ENTRENAMIENTO
#
# train_on_responses_only con los delimitadores correctos de Mistral:
#   instruction_part = "[INST] "   (inicio del turno user)
#   response_part    = "[/INST]"   (inicio del turno assistant, SIN espacio)
# El template de Mistral genera: <s>[INST] {user} [/INST]{target}</s>
# ---------------------------------------------------------------

trainer = SFTTrainer(
    model              = model,
    tokenizer          = tokenizer,
    train_dataset      = train_dataset,
    eval_dataset       = val_dataset,
    dataset_text_field = "text",
    max_seq_length     = MAX_SEQ_LENGTH,
    dataset_num_proc   = 2,
    packing            = False,
    args = SFTConfig(
        output_dir                  = "mistral7b_tu_lora",
        per_device_train_batch_size = BATCH_SIZE,
        gradient_accumulation_steps = GRAD_ACCUM,
        num_train_epochs            = NUM_EPOCHS,
        learning_rate               = LEARNING_RATE,
        warmup_steps                = WARMUP_STEPS,
        weight_decay                = 0.05,
        neftune_noise_alpha         = 5,
        fp16                        = not is_bfloat16_supported(),
        bf16                        = is_bfloat16_supported(),
        logging_steps               = 5,
        eval_strategy               = "epoch",
        save_strategy               = "epoch",
        load_best_model_at_end      = True,
        metric_for_best_model       = "eval_loss",
        greater_is_better           = False,
        optim                       = "paged_adamw_8bit",
        seed                        = SEED,
        report_to                   = "none",
    ),
    callbacks = [EarlyStoppingCallback(early_stopping_patience=PATIENCE)],
)

# Delimitadores correctos para Mistral-7B-Instruct-v0.3
# Template: <s>[INST] {user} [/INST]{assistant}</s>
trainer = train_on_responses_only(
    trainer,
    instruction_part = "[INST] ",
    response_part    = "[/INST]",
)

print("\nEmpieza entrenamiento Mistral-7B-Instruct-v0.3 (QLoRA)")
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()
TRAIN_START = time.perf_counter()
trainer.train()
total_train_time = time.perf_counter() - TRAIN_START
print(f"Entrenamiento completado en {total_train_time:.2f}s ({total_train_time/60:.2f} min)")
if device == "cuda":
    print(f"Pico de memoria: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB")

# ---------------------------------------------------------------
# PREPARAR PARA INFERENCIA
# ---------------------------------------------------------------

model.eval()
tokenizer.padding_side = "left"

# ---------------------------------------------------------------
# MÉTRICAS GEC — idénticas al prompting de Mistral (doc 3)
# ---------------------------------------------------------------

def decode_pred(token_ids):
    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=True,
    ).strip()


def clean_pred(text: str) -> str:
    text = text.strip()
    text = re.sub(
        r'^(corrección|corregido|texto corregido|output|respuesta)\s*:\s*',
        '', text, flags=re.IGNORECASE,
    )
    text = re.sub(r'^[\s\-–—\"\']+', '', text)
    text = text.strip().strip('"\'')
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
    if lines:
        text = lines[0]
    sentences = re.split(r'(?<=[.!?])\s+', text)
    return sentences[0].strip() if sentences else text.strip()


def compute_gec_metrics(src_text, tgt_text, pred_text):
    if not pred_text:
        pred_text = src_text

    src_tokens  = src_text.split()
    tgt_tokens  = tgt_text.split()
    pred_tokens = pred_text.split()

    needs_correction = src_text.strip() != tgt_text.strip()
    pred_modified    = pred_text.strip() != src_text.strip()
    gleu             = sentence_gleu([tgt_tokens], pred_tokens)

    tgt_c  = Counter(tgt_tokens)
    pred_c = Counter(pred_tokens)
    tp_tok = sum((tgt_c & pred_c).values())
    fp_tok = sum((pred_c - tgt_c).values())
    fn_tok = sum((tgt_c - pred_c).values())

    precision_tok = tp_tok / (tp_tok + fp_tok) if (tp_tok + fp_tok) > 0 else 0.0
    recall_tok    = tp_tok / (tp_tok + fn_tok) if (tp_tok + fn_tok) > 0 else 0.0
    beta    = 0.5
    f05_tok = (
        (1 + beta**2) * precision_tok * recall_tok
        / (beta**2 * precision_tok + recall_tok)
        if (beta**2 * precision_tok + recall_tok) > 0 else 0.0
    )
    ed_raw  = edit_distance(tgt_tokens, pred_tokens)
    ed_norm = ed_raw / max(len(tgt_tokens), len(pred_tokens), 1)

    return {
        "needs_correction": needs_correction,
        "pred_modified":    pred_modified,
        "gleu":             gleu,
        "f05_tok":          f05_tok,
        "precision_tok":    precision_tok,
        "recall_tok":       recall_tok,
        "ed_norm":          ed_norm,
        "ed_raw":           ed_raw,
    }


def aggregate_metrics(all_metrics):
    if not all_metrics:
        return {}
    n  = len(all_metrics)
    tp = sum(1 for m in all_metrics if     m["needs_correction"] and     m["pred_modified"])
    tn = sum(1 for m in all_metrics if not m["needs_correction"] and not m["pred_modified"])
    fp = sum(1 for m in all_metrics if not m["needs_correction"] and     m["pred_modified"])
    fn = sum(1 for m in all_metrics if     m["needs_correction"] and not m["pred_modified"])

    accuracy  = (tp + tn) / n
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    beta      = 0.5
    f05       = (
        (1 + beta**2) * precision * recall / (beta**2 * precision + recall)
        if (beta**2 * precision + recall) > 0 else 0.0
    )
    return {
        "n_total":           n,
        "n_positive":        tp + fn,
        "n_negative":        tn + fp,
        "tp": tp, "tn": tn, "fp": fp, "fn": fn,
        "accuracy":          accuracy,
        "precision":         precision,
        "recall":            recall,
        "f05":               f05,
        "gleu_mean":         float(np.mean([m["gleu"]          for m in all_metrics])),
        "gleu_std":          float(np.std( [m["gleu"]          for m in all_metrics])),
        "f05_tok_mean":      float(np.mean([m["f05_tok"]       for m in all_metrics])),
        "f05_tok_std":       float(np.std( [m["f05_tok"]       for m in all_metrics])),
        "precision_tok":     float(np.mean([m["precision_tok"] for m in all_metrics])),
        "recall_tok":        float(np.mean([m["recall_tok"]    for m in all_metrics])),
        "ed_norm_mean":      float(np.mean([m["ed_norm"]       for m in all_metrics])),
        "correction_rate":   float(sum(1 for m in all_metrics if m["pred_modified"]) / n),
        "false_corrections": fp,
    }


def print_all_samples(preds, dataset_name):
    print(f"\n{'='*60}")
    print(f"TODAS LAS MUESTRAS -- {dataset_name} ({len(preds)} total)")
    print(f"{'='*60}")
    for i, p in enumerate(preds):
        nc, pm = p["needs_correction"], p["pred_modified"]
        if   nc and     pm: label = "TP  (con errores -> modelo corrige)"
        elif nc and not pm: label = "FN  (con errores -> modelo no corrige)"
        elif not nc and pm: label = "FP  (sin errores -> modelo modifica igualmente)"
        else:               label = "TN  (sin errores -> modelo no toca)"
        print(f"\n[{i+1}/{len(preds)}] {label}")
        print(f"  Fuente:   {p['src']}")
        print(f"  Esperado: {p['tgt']}")
        print(f"  Generado: {p['pred']}")
        print(f"  GLEU={p['gleu']:.3f} | F0.5_tok={p['f05_tok']:.3f} | ED_norm={p['ed_norm']:.3f}")


def print_aggregated_metrics(agg, inference_time, n_samples, dataset_name):
    print(f"\n{'='*60}")
    print(f"METRICAS AGREGADAS -- {dataset_name}")
    print(f"{'='*60}")
    print(f"  Muestras totales:  {agg.get('n_total', 0)}")
    print(f"  Con errores (pos): {agg.get('n_positive', 0)}")
    print(f"  Sin errores (neg): {agg.get('n_negative', 0)}")
    print()
    print(f"  -- Deteccion (clasificacion binaria) --")
    print(f"     TP={agg.get('tp',0)}  TN={agg.get('tn',0)}  "
          f"FP={agg.get('fp',0)}  FN={agg.get('fn',0)}")
    print(f"     Accuracy:   {agg.get('accuracy',  0):.4f}  (frases clasificadas correctamente)")
    print(f"     Precision:  {agg.get('precision', 0):.4f}  (de las que corrige, cuantas tenian error)")
    print(f"     Recall:     {agg.get('recall',    0):.4f}  (de las que tenian error, cuantas detecta)")
    print(f"     F0.5:       {agg.get('f05',       0):.4f}  (penaliza mas FP que FN, estandar GEC)")
    print()
    print(f"  -- Calidad de correccion (token-level) --")
    print(f"     GLEU:           {agg.get('gleu_mean',    0):.4f} +/- {agg.get('gleu_std', 0):.4f}")
    print(f"     F0.5 tok:       {agg.get('f05_tok_mean', 0):.4f} +/- {agg.get('f05_tok_std', 0):.4f}")
    print(f"     Precision tok:  {agg.get('precision_tok', 0):.4f}")
    print(f"     Recall tok:     {agg.get('recall_tok',    0):.4f}")
    print(f"     ED norm:        {agg.get('ed_norm_mean',  0):.4f}")
    print()
    print(f"  -- Comportamiento del modelo --")
    print(f"     Correction rate:   {agg.get('correction_rate',   0):.4f}")
    print(f"     False corrections: {agg.get('false_corrections',  0)}")
    print()
    print(f"  -- Tiempos --")
    print(f"     Inferencia total: {inference_time:.4f}s")
    print(f"     Throughput:       {n_samples/inference_time:.2f} muestras/s")
    print(f"     ms/muestra:       {inference_time/n_samples*1000:.2f}ms")

# ---------------------------------------------------------------
# EVALUACIÓN
# ---------------------------------------------------------------

class GECDataset(TorchDataset):
    def __init__(self, df):
        self.inputs  = df["input"].tolist()
        self.targets = df["target"].tolist()
    def __len__(self):  return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

test_loader     = DataLoader(GECDataset(df_test),           batch_size=1, shuffle=False)
noticias_loader = DataLoader(GECDataset(df_noticias_final), batch_size=1, shuffle=False)
print(f"\nTest: {len(test_loader)} muestras | Noticias: {len(noticias_loader)} muestras")


def evaluate_model(loader, src_texts_list=None):
    model.eval()
    all_metrics, predictions = [], []

    with torch.no_grad():
        for i, (input_texts, tgt_texts) in enumerate(loader):
            input_text = input_texts[0]
            tgt_text   = tgt_texts[0]

            enc = tokenizer.apply_chat_template(
                build_prompt(input_text),
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
            input_ids      = enc["input_ids"].to(device)
            attention_mask = enc["attention_mask"].to(device)
            input_length   = input_ids.shape[1]

            outputs = model.generate(
                input_ids          = input_ids,
                attention_mask     = attention_mask,
                max_new_tokens     = MAX_NEW_TOKENS,
                do_sample          = False,
                num_beams          = 1,
                repetition_penalty = REPETITION_PENALTY,
                use_cache          = False,
                pad_token_id       = tokenizer.pad_token_id,
                eos_token_id       = MISTRAL_EOS_ID,
            )

            pred_raw  = decode_pred(outputs[0][input_length:])
            pred_text = clean_pred(pred_raw)
            src_text  = src_texts_list[i] if src_texts_list else input_text

            metrics = compute_gec_metrics(src_text, tgt_text, pred_text)
            all_metrics.append(metrics)
            predictions.append({
                "src": src_text, "tgt": tgt_text,
                "pred_raw": pred_raw, "pred": pred_text,
                **metrics,
            })

    return aggregate_metrics(all_metrics), predictions


# TEST
print("\n" + "=" * 60)
print("EVALUACION EN TEST SET — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()

t0 = time.perf_counter()
test_agg, test_preds = evaluate_model(test_loader, df_test["input"].tolist())
if torch.cuda.is_available():
    torch.cuda.synchronize()
test_inference_time = time.perf_counter() - t0

print_all_samples(test_preds, "TEST SET")
print_aggregated_metrics(test_agg, test_inference_time, len(test_preds), "TEST SET")

# NOTICIAS
print("\n" + "=" * 60)
print("EVALUACION EN NOTICIAS — Mistral-7B-Instruct-v0.3 (fine-tuned QLoRA)")
print("=" * 60)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

t0 = time.perf_counter()
noticias_agg, noticias_preds = evaluate_model(
    noticias_loader, df_noticias_final["input"].tolist()
)
if torch.cuda.is_available():
    torch.cuda.synchronize()
noticias_inference_time = time.perf_counter() - t0

print_all_samples(noticias_preds, "NOTICIAS")
print_aggregated_metrics(noticias_agg, noticias_inference_time, len(noticias_preds), "NOTICIAS")

# ---------------------------------------------------------------
# RESULTADOS FINALES
# ---------------------------------------------------------------

if device == "cuda":
    max_mem_gb  = torch.cuda.max_memory_allocated(0) / 1e9
    reserved_gb = torch.cuda.memory_reserved(0) / 1e9
else:
    max_mem_gb = reserved_gb = 0.0

results = {
    "model_name":          "Mistral-7B-Instruct-v0.3",
    "evaluation_mode":     "fine-tuned QLoRA",
    "seed":                SEED,
    "device":              device,
    "n_params":            n_params,
    "n_params_trainable":  n_params_train,
    "quantization":        "nf4_4bit",
    "lora_r":              LORA_R,
    "lora_alpha":          LORA_ALPHA,
    "lora_dropout":        LORA_DROPOUT,
    "epochs":              NUM_EPOCHS,
    "learning_rate":       LEARNING_RATE,
    "warmup_steps":        WARMUP_STEPS,
    "patience":            PATIENCE,
    "neftune_noise_alpha": 5,
    "batch_size":          BATCH_SIZE,
    "grad_accum":          GRAD_ACCUM,
    "effective_batch":     BATCH_SIZE * GRAD_ACCUM,
    "max_seq_length":      MAX_SEQ_LENGTH,
    "max_new_tokens":      MAX_NEW_TOKENS,
    "do_sample":           False,
    "repetition_penalty":  REPETITION_PENALTY,
    "decode_strategy":     "greedy",
    "loss_masking":        "train_on_responses_only",
    "total_train_time_s":  total_train_time,

    "test_n_total":           test_agg.get("n_total",    0),
    "test_n_positive":        test_agg.get("n_positive", 0),
    "test_n_negative":        test_agg.get("n_negative", 0),
    "test_tp":                test_agg.get("tp",  0),
    "test_tn":                test_agg.get("tn",  0),
    "test_fp":                test_agg.get("fp",  0),
    "test_fn":                test_agg.get("fn",  0),
    "test_accuracy":          test_agg.get("accuracy",  0),
    "test_precision":         test_agg.get("precision", 0),
    "test_recall":            test_agg.get("recall",    0),
    "test_f05":               test_agg.get("f05",       0),
    "test_gleu_mean":         test_agg.get("gleu_mean",     0),
    "test_gleu_std":          test_agg.get("gleu_std",      0),
    "test_f05_tok_mean":      test_agg.get("f05_tok_mean",  0),
    "test_f05_tok_std":       test_agg.get("f05_tok_std",   0),
    "test_precision_tok":     test_agg.get("precision_tok", 0),
    "test_recall_tok":        test_agg.get("recall_tok",    0),
    "test_ed_norm_mean":      test_agg.get("ed_norm_mean",  0),
    "test_correction_rate":   test_agg.get("correction_rate",   0),
    "test_false_corrections": test_agg.get("false_corrections", 0),
    "test_inference_time_s":  test_inference_time,
    "test_throughput_sps":    len(test_preds) / test_inference_time,
    "test_ms_per_sample":     test_inference_time / len(test_preds) * 1000,

    "noticias_n_total":           noticias_agg.get("n_total",    0),
    "noticias_n_positive":        noticias_agg.get("n_positive", 0),
    "noticias_n_negative":        noticias_agg.get("n_negative", 0),
    "noticias_tp":                noticias_agg.get("tp",  0),
    "noticias_tn":                noticias_agg.get("tn",  0),
    "noticias_fp":                noticias_agg.get("fp",  0),
    "noticias_fn":                noticias_agg.get("fn",  0),
    "noticias_accuracy":          noticias_agg.get("accuracy",  0),
    "noticias_precision":         noticias_agg.get("precision", 0),
    "noticias_recall":            noticias_agg.get("recall",    0),
    "noticias_f05":               noticias_agg.get("f05",       0),
    "noticias_gleu_mean":         noticias_agg.get("gleu_mean",     0),
    "noticias_gleu_std":          noticias_agg.get("gleu_std",      0),
    "noticias_f05_tok_mean":      noticias_agg.get("f05_tok_mean",  0),
    "noticias_f05_tok_std":       noticias_agg.get("f05_tok_std",   0),
    "noticias_precision_tok":     noticias_agg.get("precision_tok", 0),
    "noticias_recall_tok":        noticias_agg.get("recall_tok",    0),
    "noticias_ed_norm_mean":      noticias_agg.get("ed_norm_mean",  0),
    "noticias_correction_rate":   noticias_agg.get("correction_rate",   0),
    "noticias_false_corrections": noticias_agg.get("false_corrections", 0),
    "noticias_inference_time_s":  noticias_inference_time,
    "noticias_throughput_sps":    len(noticias_preds) / noticias_inference_time,
    "noticias_ms_per_sample":     noticias_inference_time / len(noticias_preds) * 1000,

    "gpu_max_memory_gb":      max_mem_gb,
    "gpu_reserved_memory_gb": reserved_gb,
}

print("\n" + "=" * 60)
print("RESULTADOS FINALES -- RESUMEN")
print("=" * 60)
for k, v in results.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

with open("results_mistral7b_finetuned.json", "w", encoding="utf-8") as f:
    json.dump({
        "summary":              results,
        "test_predictions":     test_preds,
        "noticias_predictions": noticias_preds,
    }, f, indent=2, ensure_ascii=False)

print("\nResultados guardados en 'results_mistral7b_finetuned.json'")
```
